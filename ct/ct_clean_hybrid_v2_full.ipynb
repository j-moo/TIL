# 컴퓨터 사고력(CT) 필기 정리본

이 노트는 **내가 필기한 내용(스크린샷)**을 최대한 유지하면서,
읽기 좋게 흐름을 잡고(개념 → 예시 → 확인 코드), 필요한 설명을 덧붙인 버전입니다.

- 구성: 목표 → 개념 → 예시/문제 → 확인 코드 → (필요 시) 추가 설명 → 요약/미니퀴즈
- 원본을 줄이지 않기 위해 **각 파트에 관련 스크린샷을 그대로 첨부**했습니다.


## 오늘 범위/목표

- **명제와 논리연산**(부정, AND/OR/XOR, 조건명제, 쌍조건명제) 이해하기
- 조건명제의 **역/이/대우**를 구분하고, 참/거짓 판단 기준 익히기
- **진리표를 직접 만드는 순서**를 익히기
- `∀x, P(x)→Q(x)` 형태에서 자주 나오는 **Trivial / Vacuous proof** 감 잡기
- **재귀(recursion)**를 “호출→복귀” 흐름으로 설명할 수 있게 만들기
- 대표 점화식들을 전개(unfolding)해서 **시간복잡도 Big-O**로 정리하기

## 0. 서론
- 명제 : 참과 거짓으로 판별할 수 있는 문장
  - 2는 짝수이다. -> 참인 명제
  - 3은 짝수이다. -> 거짓인 면제
  - 서울은 대한민국의 수도이다. -> 참인 명제
  - 강의는 재밌다. -> 명제가 아니다.
  - 강사님은 멋있다. -> 명제가 아니다.

- 명제는 보통 이런식으로 많이 사용된다.
  - p -> q (조건명제)
    - p가 참일 때, q도 참이다.

- 예시
- 4가 짝수일 때, 4^2 도 짝수이다.

### 명제를 증명하는 방법

#### 1. 대우 증명
- 대우명제: 4^2이 홀수라면, 4는 홀수이다.
  - 가정이 거짓 -> 무조건 참이다.
  - 대우명제가 참이므로, 원래 명제도 참이다.

#### 2. 직접 증명
- 4가 짝수다 = 4 = (2k)
 -> 4^2 = (2k)^2 = 4(k^2)
 -> 4(k^2) 은 4로 나누어 떨어진다 => 짝수다.

- 귀납법 (대표적인 직접 증명)
  - 특정 사례로 부터 출발해서, 일반적인 규칙을 찾는 방법
    - 기저 사례
    - 귀납 가정
    - 귀납 단계
    - 하나가 되면 끝까지 된다.
  - 도미노 예시
    - 모든 도미노는 연속으로 쓰러진다.
      - 기초 단계: 첫 번째 도미노를 손으로 민다 -> 쓰러짐
      - 귀납 가정: 임의의 자연수 n에 대해, n번째 도미노가 쓰러졌다고 가정하자. 즉, P(n)이 참이라고 가정한다.
      - 귀납 단계: n번째 도미노가 쓰러졌다고 하자.
        - (즉, 𝑃(𝑛)이 참이라고 가정)
        - 도미노는 충분히 가깝게 배치되어 있어서 하나가 쓰러지면 반드시 다음 것도 쓰러진다.
        - 따라서 n번째가 쓰러지면 -> n+1번째도 쓰러진다.
      - 결론: 1번이 쓰러지고 쓰러진 것은 다음 것을 쓰러뜨리므로 모든 도미노가 쓰러진다.
  - [주의] 경험적 귀납법 예시
    - 명제: n일 연속으로 하루 1번 더 연락하면, 상대의 호감도는 n단계 올라간다.
    - 기초 단계: 첫째 날, 평소보다 한 번 더 연락 -> 호감도 상승 (1단계 상승)
      - 즉, P(1)이 참이라고 관찰했다.
    - 귀납 가정: 임의의 n일 동안 매일 한 번 더 연락하면 호감도가 n단계 올라갔다고 가정하자. 즉, P(n)이 참이라고 가정한다.
    - 귀납 단계
      - 첫째 날, 평소보다 한 번 더 연락 -> 호감도 상승
      - 둘째 날, 평소보다 한 번 더 연락 -> 호감도 상승
      - n일 연속 연락 -> n일 연속 호감도 상승

#### 3. 간접 증명 (모순 증명)
- 반대 명제 : 4^2 이 홀수라고 가정하자.
  -> 4^2 = 4(k^2) -> 이건 짝수인데 ...?
  -> 반대명제은 모순이 발생하구나!
  -> 원래 명제가 참이다.

논리의 정의

논리(Hard Logic)
생각이나 주장의 타당성과 옳고 그름을 판별하는 규칙이나 원리를 의미합니다. 경험적, 직관적, 확률적 판단을 의미하는 Soft Logic에 반대되는 의미로 Hard Logic이라고 표현합니다. 맞는 이유와 틀린 이유를 정확히 밝히는 생각의 규칙입니다.

프로그래밍의 어려운 점

카드 문제와 맥주집 문제의 비교
- 맥주집 문제가 훨씬 풀기 쉬움
  - 사실, 두 문제는 완전히 같은 문제입니다. 즉, 논리적 구성이 완전히 동일합니다.
- 왜 맥주집 문제가 풀기 쉬운가?
  - 논리 구조를 정확히 이해하고 맥주집 문제를 푸는 사람은 카드 문제를 똑같이 풀 수 있습니다.
  - 맥주집 문제를 풀 때 논리를 사용한 것이 아닙니다.

Soft Logic
- 맥주집 문제를 풀 때 직관을 사용한 것과 같은 경우
- 직관은 논리적인 느낌을 주는 것
- 직관의 장점은 익숙한 상황에서 빠름
- 직관의 단점은 정확하지 않음 (가끔은 익숙한 상황에서도 틀림)
- 또 다른 단점은 강한 착각을 일으킨다는 것

Hard Logic VS Soft Logic
- 일상 생활
  - Soft Logic이 빠르기 때문에 유용합니다.
  - 논리적으로 부정확한 표현을 사용하지만, 어떤 의미인지 모든 사람이 이미 알고 있다는 가정이 존재합니다.
- 프로그래밍
  - 프로그래밍 언어의 표현들이 모두 논리학에서 나온 것입니다.
  - 사용되는 수많은 알고리즘들을 이해하기 위해서는 Hard Logic이 필요합니다.
- 문제에 대해 오해하는 경우
  - Soft Logic으로 알고리즘을 이해하려고 하는 경우입니다.
  - 알고리즘 설명을 보고 또 봐도 이해가 안되는 것은 증명을 안 봤기 때문입니다.
  - 증명을 봐도 이해가 안되는 것은 직관으로 이해하려고 하기 때문입니다.
  - 가끔 직관적으로 이해되는 알고리즘이 있지만, 조금만 어려워지면 직관으로 완전한 이해를 얻는 것은 사실상 불가능합니다.

명제

명제의 연산(결합)
- 부정 (NOT)
- p가 명제일 떄, 명제의 진릿값이 반대가 됩니다.
- ~p 또는 -p로 표기하고, not p 또는 p의 부정으로 읽습니다.

논리곱(AND)
- p, q가 명제일 때, p, q 모두 참일 때만 참이 되는 명제입니다.
- p^q로 표기하고, p and q 또는 p 또는 q로 읽습니다.

명제의 연산(결합)
- 논리합 (OR)
- p, q가 명제일 때, p, q 모두 거짓일 때만 거짓이 되는 명제입니다.
- p V q로 표기하고 p or q나 p 또는 q로 읽습니다.

명제의 연산(결합)
- 배타적 논리합 (XOR)
  - p, q가 명제일 때, p, q중 하나만 참일 때 참이 되는 명제입니다.
  - p + q로 표기하고, p xor q 또는 p exclusive OR q로 읽습니다.

조건명제
- p, q가 명제일 때, 명제 p가 조건(원인), q가 결론(결과)으로 제시되는 명제
- p -> q (p이면 q이다)

쌍방조건명제
- p, q가 명제일 때, 명제 p와 q가 모두 조건이면서 결론인 명제
- p <-> q (p면 q고, q면 p다)

조건명제의 역, 이, 대우
- 주어진 명제 : p -> q
- 역 : q -> p
- 이 : ~p -> ~q
- 대우: ~q -> ~p

쉽게 말하면,
- `p → q`는 “약속” 같은 거라서 **약속을 깬 경우( p 참, q 거짓 )**만 거짓이에요.


### 확인 코드: 진리표를 파이썬으로 직접 찍어보기

````python
from itertools import product

def tf(b): 
    return 'T' if b else 'F'

def truth_table_binary(op_name, op):
    print(f'=== {op_name} ===')
    print('p q | result')
    print('---------')
    for p,q in product([True, False],[True, False]):
        print(tf(p), tf(q), '|', tf(op(p,q)))
    print()

truth_table_binary('p AND q', lambda p,q: p and q)
truth_table_binary('p OR q', lambda p,q: p or q)
truth_table_binary('p XOR q', lambda p,q: (p and not q) or (not p and q))
truth_table_binary('p -> q', lambda p,q: (not p) or q)
truth_table_binary('p <-> q', lambda p,q: p == q)
````

## 4. 연습문제: `p → q`가 거짓일 때

`p → q`가 **거짓**이 되려면 딱 1가지 경우뿐입니다.

- `p = 참`, `q = 거짓`

이 조건만 잡고 나머지를 계산하면 돼요.

![문제 2](img_08.png)

결론(필기 풀이 요지):
1) `~p → q` : `~p`는 거짓, 그래서 조건명제는 참  
2) `p ∨ q` : `p`가 참이므로 참  
3) `q → p` : `q`가 거짓이므로 조건명제는 참


### 확인 코드

````python
# 확인: p->q가 거짓인 케이스는 (p=True, q=False) 뿐
p, q = True, False

def imp(a,b): return (not a) or b

print("p,q =", p,q)
print("p->q:", imp(p,q))
print("~p->q:", imp(not p, q))
print("p OR q:", p or q)
print("q->p:", imp(q, p))
````

## 6. 증명 패턴: Trivial / Vacuous Proof

### 6-1) Trivial Proof
`∀x, P(x) → Q(x)`를 증명하려고 할 때, **Q(x)가 항상 참이면** 전체가 자동으로 참이 됩니다.

![Trivial Proof 예시](img_12.png)

- 왜냐하면 조건명제는 `Q(x)`가 참이면 (p가 뭐든) 결과가 참이기 때문

### 6-2) Vacuous Proof
`∀x, P(x) → Q(x)`에서 **P(x)가 항상 거짓이면** 역시 전체가 자동으로 참입니다.

![Vacuous Proof 예시](img_13.png)

- 왜냐하면 조건명제는 `P(x)`가 거짓이면 (q가 뭐든) 참이기 때문

👉 둘 다 결론은 “조건명제의 진리표 성질”에서 나옵니다.


## 7. 재귀(Recursion): 호출과 복귀를 같이 보기

재귀는 결국 “함수가 자기 자신을 다시 부르는 것”인데,
중요한 건 2가지입니다.

1) **기저 조건(base case)**: 언제 멈추는가?  
2) **문제를 줄이는 규칙**: 다음 호출은 무엇인가?

### 예시: 1부터 x까지 합
![sum(x) 코드](img_14.png)

이 코드는:
- `x <= 0`이면 0을 리턴하면서 멈추고
- 아니면 `x + sum(x-1)`로 한 단계 작아진 문제를 다시 부릅니다.


### 코드로 직접 실행해보기

````python
def sum_rec(x):
    if x <= 0:
        return 0
    return x + sum_rec(x-1)

for n in range(0,6):
    print(n, sum_rec(n))
````

### 호출이 어떻게 풀리는지(필기 흐름)

- `sum(1)`은 결국 `1 + sum(0)`이 되고, `sum(0)`이 0이라서 끝납니다.  
![sum(1) 전개](img_15.png)

- 일반화해서 `sum(k+1) = (k+1) + sum(k)` 형태로 생각할 수도 있어요.  
![sum(k+1) 전개](img_16.png)

### “멈추지 않는 재귀”도 같이 봐야 함
![기저 조건 없는 함수 vs 있는 함수](img_18.png)

- `abc(x)`는 기저 조건이 없어서 끝이 없음  
- `sum(x)`는 `x<=0`에서 멈춤

### 호출→복귀(리턴 값이 쌓이는 느낌)
![sum(4) 호출/복귀 흐름](img_19.png)

## 8. 점화식(Recurrence) 전개로 시간복잡도 구하기

여기서는 대표적으로 `T(n)=...` 형태를 **반복해서 풀어쓰는(unfolding)** 방식으로 정리합니다.

### 8-1) `T(n)=T(n-1)+1`
![문제1](img_20.png)

### 8-2) `T(n)=T(n-1)+n`
![문제2](img_21.png)

### 8-3) `T(n)=T(n-1)+log n`
![문제3](img_22.png)

이런 형태는 보통:
- `+1`이면 `O(n)`
- `+n`이면 `O(n^2)`
- `+log n`이면 `O(n log n)`  
같은 식으로 “더해지는 항”의 누적을 보면 감이 옵니다.


### 8-4) `T(n)=T(n/2)+1` (과제)
![문제4](img_23.png)

힌트: `n → n/2 → n/4 → ... → 1`  
몇 번 나누면 1이 될까? → **log₂ n 번**

그래서 보통 `O(log n)`이 됩니다.

### 8-5) `T(n)=T(n/2)+n`
![문제5](img_24.png)

### 8-6) `T(n)=2T(n/2)+n`
![문제6](img_25.png)

이건 병합정렬(Merge Sort) 같은 데서 등장하는 대표 형태고, 보통 `O(n log n)`입니다.

### 8-7) `T(n)=3T(n/2)+n`
![문제7](img_26.png)

`n^{log₂ 3}` 형태가 나오면서, 분할정복에서 “가지 수”가 늘어나면 지수가 커진다는 걸 보여줍니다.

### 8-8) `T(n)=T(n-1)+1/n`
![문제8](img_27.png)

`1 + 1/2 + 1/3 + ... + 1/n` (조화급수) 느낌이라서 대략 `O(log n)`으로 갑니다.


### 확인 코드: 점화식을 실제 값으로 찍어보기(감 잡기)

````python
import math

def T1(n):
    # T(n)=T(n-1)+1, T(0)=1
    t=1
    for k in range(1,n+1):
        t = t + 1
    return t

def T2(n):
    # T(n)=T(n-1)+n, T(0)=1
    t=1
    for k in range(1,n+1):
        t = t + k
    return t

def T4(n):
    # T(n)=T(n//2)+1, T(1)=1 (정수 n 가정)
    t=1
    while n>1:
        n//=2
        t += 1
    return t

def T6(n):
    # T(n)=2T(n/2)+n, T(1)=1 (n이 2의 거듭제곱이라고 가정)
    # 전개 대신 값 계산용
    if n==1: return 1
    return 2*T6(n//2) + n

for n in [1,2,4,8,16,32,64]:
    print(n, T1(n), T2(n), T4(n), T6(n))
````

## 9. 요약(핵심만 한 번 더)

- **조건명제 `p→q`는 ‘p 참, q 거짓’일 때만 거짓**  
  → 그래서 “반례 찾기” 문제가 많음(카드/맥주집)
- `p→q`와 **대우 `~q→~p`는 동치** (참/거짓이 항상 같다)
- 진리표는 **작은 항부터 열을 만들어** 조립한다
- `∀x, P(x)→Q(x)`에서  
  - Q(x)가 항상 참이면 **Trivial proof** 느낌으로 자동 참  
  - P(x)가 항상 거짓이면 **Vacuous proof** 느낌으로 자동 참
- 재귀는 **기저 조건 + 문제 크기 줄이기**가 핵심, 호출과 복귀를 함께 본다
- 점화식은 전개해서 더해지는 항이 어떻게 누적되는지 보면 Big-O가 나온다
