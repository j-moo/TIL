읽기 좋게 흐름을 잡고(개념 → 예시 → 확인 코드), 필요한 설명을 덧붙인 버전입니다.

## 0. 왜 ‘논리’를 배우나?

프로그래밍이 어려운 이유는 크게 2가지가 섞여 있기 때문입니다.

1) **문법/라이브러리**  
- 처음엔 낯설어서 어렵고, 연습하면 익숙해지는 부분

2) **논리(사고의 규칙)**  
- “그럴 것 같아”는 빠르지만, 상황이 바뀌면 쉽게 틀릴 수 있음  
- **왜 맞는지 / 왜 틀린지**를 설명할 수 있어야 안정적으로 코드를 짤 수 있음

여기서 자주 나오는 구분이 있습니다.

- **Hard Logic**: 맞는 이유/틀린 이유를 정확히 밝히는 “생각의 규칙”
- **Soft Logic(직관)**: 익숙한 상황에서는 빠르지만, 착각을 만들 수 있음

쉽게 말하면,  
문법은 “키보드 타이핑”이라면, 논리는 “지도 읽기”에 가깝습니다.  
아래 카드/맥주집 문제는 “어떤 것만 확인하면 규칙을 깰 수 있는지”를 찾는 연습입니다.

### 카드 문제(와슨 선택 과제 느낌)

> 사실: 카드 한쪽은 알파벳, 다른 쪽은 숫자  
> 주장: **한쪽이 D이면 반대쪽은 3이다**

아래 카드 중 **반드시 뒤집어 확인해야 하는 카드**는 무엇일까?

![카드 문제](images/img_00.png)

핵심은 “주장을 **깨뜨릴 수 있는 경우(반례)**를 찾는 것”입니다.

- `D`를 뒤집는다 → D 뒤가 3이 아니면 주장이 즉시 깨짐 ✅  
- `7`을 뒤집는다 → 7 뒤가 D이면 “D인데 3이 아님”이 되어 주장이 깨짐 ✅  
- `3`은 굳이 볼 필요가 없음: 3 뒤가 D이든 아니든 “D이면 3”을 깨지 못함  
  (주장은 **D인 경우만** 제한을 걸기 때문)

👉 그래서 보통 답은 **D와 7**입니다.


### 맥주집 문제(조건을 깨는 경우 찾기)

규칙: **20세 이하인 사람은 맥주를 마실 수 없음**

이런 문제도 마찬가지로 “규칙을 깨는 경우”를 찾으면 돼요.

- **맥주를 마시는 사람**이 20세 이하인지 확인해야 함  
- **20세 이하**인 사람이 맥주를 마시는지 확인해야 함

(반대로, 콜라를 마시는 25세를 확인해도 규칙 위반 여부와 상관이 없어요.)


## 1. 명제(Proposition)와 참/거짓

- **명제**: 참(True) 또는 거짓(False)으로 판단 가능한 문장  
  예) “2는 짝수이다”(참), “3은 짝수이다”(거짓)

- **명제가 아닌 것**: 기준이 사람마다 달라서 참/거짓을 확정 못 하는 문장  
  예) “강의는 재밌다”, “강사님은 멋있다” (감상/의견이라 판정 불가)

명제는 보통 기호로:
- `p`, `q` 처럼 한 문장을 대표시키고,
- 연산을 붙여서 복합 명제를 만듭니다.

참/거짓은 상황에 따라 `T/F`로 쓰기도 하고, 코딩에서는 `1/0`으로 생각해도 됩니다.

아래 표들은 “p, q가 참/거짓일 때 결과가 어떻게 되는지”를 한 번에 보여줍니다.

### 2-2) AND: `p ∧ q`
(둘 다 참일 때만 참)
![AND 진리표](images/img_01.png)

### 2-3) OR: `p ∨ q`
(하나라도 참이면 참)
![OR 진리표](images/img_02.png)

### 2-4) XOR: `p ⊕ q`
(서로 다를 때만 참)
![XOR 진리표](images/img_03.png)

### 2-5) 조건명제: `p → q`
`p`가 참인데 `q`가 거짓일 때만 **거짓**입니다. (나머지는 모두 참)
![조건명제 진리표](images/img_04.png)

### 2-6) 쌍조건명제(동치): `p ↔ q`
둘의 참/거짓이 같으면 참
![쌍조건명제 진리표](images/img_05.png)

쉽게 말하면,
- `p → q`는 “약속” 같은 거라서 **약속을 깬 경우( p 참, q 거짓 )**만 거짓이에요.


### 확인 코드: 진리표를 파이썬으로 직접 찍어보기

```
from itertools import product

def tf(b): 
    return 'T' if b else 'F'

def truth_table_binary(op_name, op):
    print(f'=== {op_name} ===')
    print('p q | result')
    print('---------')
    for p,q in product([True, False],[True, False]):
        print(tf(p), tf(q), '|', tf(op(p,q)))
    print()

truth_table_binary('p AND q', lambda p,q: p and q)
truth_table_binary('p OR q', lambda p,q: p or q)
truth_table_binary('p XOR q', lambda p,q: (p and not q) or (not p and q))
truth_table_binary('p -> q', lambda p,q: (not p) or q)
truth_table_binary('p <-> q', lambda p,q: p == q)
```

## 3. 조건명제의 ‘역/이/대우’ (엄청 자주 나옴)

조건명제는 `p → q` 형태입니다.  
진리표 기준으로는 **p가 참이고 q가 거짓일 때만 거짓**이고, 나머지는 참입니다.

![조건명제/쌍조건 진리표](images/img_04.png)
![역/이/대우 진리표 비교](images/img_06.png)
![역/이/대우 쓰기 연습](images/img_09.png)

### 용어 정리(한 번에)
- 원래 명제: `p → q`
- **역(converse)**: `q → p`
- **이(inverse)**: `~p → ~q`
- **대우(contrapositive)**: `~q → ~p`

핵심은 이것 하나:
- **원래 명제와 대우는 동치(항상 같이 참/거짓)**  
- 역/이는 동치가 보장되지 않음 (따로 판단해야 함)

#### 자주 하는 실수 포인트
- `p → q`에서 “p가 거짓이면 무조건 거짓”이라고 착각하기  
  → 실제로는 **p가 거짓이면 조건명제는 참**(약속을 적용할 상황이 아니라서)

쉽게 말하면,
- “비가 오면 우산을 쓴다”라는 약속에서  
  비가 **안 오면**(p 거짓) 우산을 썼든 안 썼든 약속을 어긴 건 아니잖아. 그게 `p 거짓이면 참` 직관이야.


## 4. 연습문제: `p → q`가 거짓일 때

`p → q`가 **거짓**이 되려면 딱 1가지 경우뿐입니다.

- `p = 참`, `q = 거짓`

이 조건만 잡고 나머지를 계산하면 돼요.

![문제 2](images/img_08.png)

결론(필기 풀이 요지):
1) `~p → q` : `~p`는 거짓, 그래서 조건명제는 참  
2) `p ∨ q` : `p`가 참이므로 참  
3) `q → p` : `q`가 거짓이므로 조건명제는 참


### 확인 코드

```
# 확인: p->q가 거짓인 케이스는 (p=True, q=False) 뿐
p, q = True, False

def imp(a,b): return (not a) or b

print("p,q =", p,q)
print("p->q:", imp(p,q))
print("~p->q:", imp(not p, q))
print("p OR q:", p or q)
print("q->p:", imp(q, p))
```

## 6. 증명(기본 도구) + Trivial / Vacuous Proof

증명은 “참이라고 말할 수 있는 근거를 만드는 과정”입니다.  
자주 쓰는 도구는 아래 4가지입니다.

- **직접 증명**: `p`를 가정하고 `q`를 보임
- **대우 증명**: `p → q` 대신 `~q → ~p`를 증명 (동치라서 가능)
- **귀류법(모순)**: 반대라고 가정해서 모순이 나옴을 보임
- **수학적 귀납법**
  - 기저 사례(Base)
  - 귀납 가정(Assumption)
  - 귀납 단계(n → n+1)

그리고 오늘 범위에서 자주 나오는 특수 패턴이 Trivial/Vacuous 입니다.

### 6-1) Trivial Proof
`∀x, P(x) → Q(x)`에서 **Q(x)가 항상 참이면** 전체가 자동으로 참입니다.

![Trivial Proof 예시](images/img_12.png)

- 조건명제는 “결론(q)이 참이면” 앞부분(p)이 뭐든 전체가 참

### 6-2) Vacuous Proof
`∀x, P(x) → Q(x)`에서 **P(x)가 항상 거짓이면** 전체가 자동으로 참입니다.

![Vacuous Proof 예시](images/img_13.png)

- 조건명제는 “가정(p)이 거짓이면” 결론(q)이 뭐든 전체가 참

👉 둘 다 결론은 “조건명제 진리표 성질”에서 나옵니다.

### 6-3) 집합/조합 메모(시험에서 자주 쓰는 문장)

- 부분집합 `A ⊆ B`를 보일 때  
  → “A의 임의의 원소 x를 잡으면, x는 B에도 들어간다”를 쓰면 깔끔합니다.

- 두 집합이 같다 `A = B`를 보일 때  
  → `A ⊆ B`와 `B ⊆ A`를 **둘 다** 보이면 됩니다.

- 조합 표기  
  - `nCk` 또는 `\(\binom{n}{k}\)`  
  - “n개 중 k개를 고르는 경우의 수”

## 7. 재귀(Recursion): 호출과 복귀를 같이 보기

재귀는 결국 “함수가 자기 자신을 다시 부르는 것”인데,
중요한 건 2가지입니다.

1) **기저 조건(base case)**: 언제 멈추는가?  
2) **문제를 줄이는 규칙**: 다음 호출은 무엇인가?

### 예시: 1부터 x까지 합
![sum(x) 코드](images/img_14.png)

이 코드는:
- `x <= 0`이면 0을 리턴하면서 멈추고
- 아니면 `x + sum(x-1)`로 한 단계 작아진 문제를 다시 부릅니다.


### 코드로 직접 실행해보기

```
def sum_rec(x):
    if x <= 0:
        return 0
    return x + sum_rec(x-1)

for n in range(0,6):
    print(n, sum_rec(n))
```

### 호출이 어떻게 풀리는지(필기 흐름)

- `sum(1)`은 결국 `1 + sum(0)`이 되고, `sum(0)`이 0이라서 끝납니다.  
![sum(1) 전개](images/img_15.png)

- 일반화해서 `sum(k+1) = (k+1) + sum(k)` 형태로 생각할 수도 있어요.  
![sum(k+1) 전개](images/img_16.png)

### “멈추지 않는 재귀”도 같이 봐야 함
![기저 조건 없는 함수 vs 있는 함수](images/img_18.png)

- `abc(x)`는 기저 조건이 없어서 끝이 없음  
- `sum(x)`는 `x<=0`에서 멈춤

### 호출→복귀(리턴 값이 쌓이는 느낌)
![sum(4) 호출/복귀 흐름](images/img_19.png)

쉽게 말하면,
- 내려갈 때는 “할 일(덧셈)이 예약”되고
- 올라올 때 그 예약이 하나씩 처리되면서 최종값이 만들어집니다.


## 8. 점화식(Recurrence) 전개로 시간복잡도 구하기

여기서는 대표적으로 `T(n)=...` 형태를 **반복해서 풀어쓰는(unfolding)** 방식으로 정리합니다.

### 8-0) 로그/비트 감각(짧은 메모)
- k비트로 표현 가능한 값의 개수는 대략 `2^k`개
- 어떤 값 `n`을 표현하려면 필요한 비트 수는 대략 `log2(n)`  
  (컴퓨터 과학에서 로그 밑은 보통 2로 생각)

### 8-1) `T(n)=T(n-1)+1`
![문제1](images/img_20.png)

### 8-2) `T(n)=T(n-1)+n`
![문제2](images/img_21.png)

### 8-3) `T(n)=T(n-1)+log n`
![문제3](images/img_22.png)

이런 형태는 보통:
- `+1`이면 `O(n)`
- `+n`이면 `O(n^2)`
- `+log n`이면 `O(n log n)`  
같은 식으로 “더해지는 항”의 누적을 보면 감이 옵니다.

힌트: `n → n/2 → n/4 → ... → 1`  
몇 번 나누면 1이 될까? → **log₂ n 번**

그래서 보통 `O(log n)`이 됩니다.

### 8-8) `T(n)=T(n-1)+1/n`
![문제8](images/img_27.png)

`1 + 1/2 + 1/3 + ... + 1/n` (조화급수) 느낌이라서 대략 `O(log n)`으로 갑니다.


### 확인 코드: 점화식을 실제 값으로 찍어보기(감 잡기)

```
import math

def T1(n):
    # T(n)=T(n-1)+1, T(0)=1
    t=1
    for k in range(1,n+1):
        t = t + 1
    return t

def T2(n):
    # T(n)=T(n-1)+n, T(0)=1
    t=1
    for k in range(1,n+1):
        t = t + k
    return t

def T4(n):
    # T(n)=T(n//2)+1, T(1)=1 (정수 n 가정)
    t=1
    while n>1:
        n//=2
        t += 1
    return t

def T6(n):
    # T(n)=2T(n/2)+n, T(1)=1 (n이 2의 거듭제곱이라고 가정)
    # 전개 대신 값 계산용
    if n==1: return 1
    return 2*T6(n//2) + n

for n in [1,2,4,8,16,32,64]:
    print(n, T1(n), T2(n), T4(n), T6(n))
```

## 9. 요약(핵심만 한 번 더)

- **조건명제 `p→q`는 ‘p 참, q 거짓’일 때만 거짓**  
  → 그래서 “반례 찾기” 문제가 많음(카드/맥주집)
- `p→q`와 **대우 `~q→~p`는 동치** (참/거짓이 항상 같다)
- 진리표는 **작은 항부터 열을 만들어** 조립한다
- `∀x, P(x)→Q(x)`에서  
  - Q(x)가 항상 참이면 **Trivial proof** 느낌으로 자동 참  
  - P(x)가 항상 거짓이면 **Vacuous proof** 느낌으로 자동 참
- 재귀는 **기저 조건 + 문제 크기 줄이기**가 핵심, 호출과 복귀를 함께 본다
- 점화식은 전개해서 더해지는 항이 어떻게 누적되는지 보면 Big-O가 나온다
