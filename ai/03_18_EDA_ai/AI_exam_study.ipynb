# 1차시

## 1. AI, ML, DL의 정의

### AI(Artificial Intelligence)
AI는 사람의 지능적인 능력을 컴퓨터가 수행하도록 만드는 넓은 분야입니다.  
여기서 지능적인 능력이란 단순 계산이 아니라, **판단, 인식, 예측, 의사결정, 이해** 같은 기능을 말합니다.

예를 들면 다음과 같습니다.
- 사진 속 사물이 고양이인지 사람인지 구분하기
- 사람 목소리를 텍스트로 바꾸기
- 번역하기
- 사용자가 좋아할 상품 추천하기

즉, AI는 **사람처럼 생각하고 판단하는 기능을 기계에 부여하려는 전체 분야**라고 볼 수 있습니다.

### ML(Machine Learning)
머신러닝은 AI를 구현하는 방법 중 하나입니다.  
전통적인 방식은 사람이 규칙을 직접 다 적어줘야 했습니다.  
하지만 현실 문제는 규칙이 너무 많고 복잡해서 사람이 전부 작성하기 어렵습니다.

그래서 머신러닝은
- 데이터를 많이 주고
- 정답을 함께 보여주거나 패턴을 찾게 해서
- 컴퓨터가 규칙을 스스로 학습하게 만드는 방식입니다.

즉, 머신러닝은 **데이터로부터 규칙을 학습하는 AI 방법론**입니다.

### DL(Deep Learning)
딥러닝은 머신러닝의 한 분야로, **여러 층의 신경망(Deep Neural Network)** 을 사용합니다.  
입력이 여러 층을 거치며 점점 더 복잡한 특징을 학습합니다.

예를 들어 이미지 인식에서는
- 처음 층: 선, 가장자리
- 중간 층: 눈, 코, 입
- 뒤쪽 층: 얼굴 전체  
처럼 점진적으로 더 고차원적인 특징을 학습할 수 있습니다.

### 포함 관계
- DL ⊂ ML ⊂ AI

### 핵심 이해 포인트
- AI는 가장 넓은 개념입니다.
- ML은 데이터를 이용해 학습하는 방법입니다.
- DL은 신경망을 깊게 쌓아 학습하는 ML의 한 종류입니다.

### 헷갈리기 쉬운 포인트
AI와 ML과 DL은 서로 완전히 다른 것이 아니라 **포함 관계**입니다.  
시험에서는 이 관계를 정확히 구분하는 문제가 자주 나옵니다.

## 2. 데이터 구성 요소

머신러닝 데이터는 보통 **입력과 정답**으로 나눠 생각합니다.

### Feature
Feature는 모델이 예측할 때 참고하는 입력 정보입니다.  
다른 말로는 **설명 변수**, **입력 변수**라고도 볼 수 있습니다.

예:
- 키
- 몸무게
- 집 면적
- 이메일 본문 길이

### Label
Label은 모델이 맞혀야 하는 정답입니다.  
지도학습에서는 학습 데이터에 Label이 함께 들어 있습니다.

예:
- 실제 집값
- 스팸 여부
- 합격/불합격
- 고양이/강아지

### 예시로 이해하기
집값 예측 문제:
- Feature: 면적, 방 개수, 위치
- Label: 실제 집값

스팸 분류 문제:
- Feature: 제목 단어, 본문 길이, 특정 단어 포함 여부
- Label: 스팸 / 정상

### 핵심 이해 포인트
- Feature는 입력값입니다.
- Label은 정답값입니다.
- 모델은 Feature를 보고 Label을 예측하도록 학습합니다.

### 헷갈리기 쉬운 포인트
집값 예측에서 **집값은 Feature가 아니라 Label**입니다.  
무엇이 입력이고 무엇이 정답인지 구분하는 문제가 자주 나옵니다.

## 3. 회귀(Regression)

회귀는 **연속형 수치값**을 예측하는 문제입니다.

### 예시
- 집값 예측
- 내일 기온 예측
- 매출 예측
- 몸무게 예측

회귀의 핵심은 정답이 **숫자 하나 또는 연속적인 값**이라는 점입니다.

### 분류와 비교
- 회귀: 72.4점, 18.3도, 3억 2천 같은 숫자를 예측
- 분류: 합격/불합격, 스팸/정상, 고양이/강아지처럼 범주를 예측

### 왜 중요한가
문제를 보았을 때 이것이 회귀인지 분류인지 구분해야  
어떤 모델과 어떤 평가 지표를 써야 하는지 정할 수 있습니다.

### 핵심 이해 포인트
회귀는 입력 데이터를 바탕으로 **연속적인 수치형 출력값**을 예측하는 문제입니다.

### 헷갈리기 쉬운 포인트
“로지스틱 회귀”는 이름에 회귀가 들어가지만 실제론 분류 문제에 많이 사용됩니다.  
그래서 시험에서는 일반 회귀와 로지스틱 회귀를 구분할 줄 알아야 합니다.

일반(선형) 회귀는 연속형 숫자를 예측하고, 로지스틱 회귀는 분류를 위한 확률을 예측하므로 둘을 구분해야 한다.

## 4. 회귀 설명력: R²(결정계수)

회귀 설명력: 회귀 모델이 실제 데이터를 얼마나 잘 설명하고 있느냐 
(이 모델이 정답에 얼마나 잘 맞추고 있나?)

R²는 회귀 모델이 데이터를 얼마나 잘 설명하는지 나타내는 대표적인 지표입니다. 
(회귀 모델이 실제 데이터의 변동을 얼마나 설명하는지 나타내는 값)

### 개념
회귀 모델은 실제값을 완벽하게 맞히기 어렵습니다.  
그래서 모델이 전체 데이터의 변동을 얼마나 설명했는지를 보는 것이 중요합니다.  
이때 사용하는 값이 R²입니다.

회귀 모델은 데이터의 패턴을 학습해 값을 예측하지만, 실제 데이터에는 잡음, 누락 변수, 측정 오차, 복잡한 비선형 관계가 존재하므로 실제값을 완벽하게 맞히기 어렵다.

### 공식
R² = 1 - RSS / TSS

- RSS: 모델이 설명하지 못한 오차의 크기
- TSS: 원래 데이터가 가진 전체 변동

### 해석
- R² = 1: 완벽한 예측
- R² = 0: 평균으로 예측한 것과 비슷한 수준
- 일반적으로 값이 클수록 설명력이 높음

예를 들어 R² = 0.8이면  
데이터 변동의 약 80%를 모델이 설명한다고 볼 수 있습니다.

### 왜 중요한가
회귀 문제에서는 단순히 오차만 보는 것뿐 아니라  
모델이 전체 구조를 얼마나 설명하는지도 중요합니다.

### 핵심 이해 포인트
R²는 **회귀 설명력**을 나타내는 지표입니다.

### 헷갈리기 쉬운 포인트
- R²는 주로 **회귀**에서 사용합니다.
- 분류 문제의 정확도와는 다른 개념입니다.

## 5. 혼동행렬(Confusion Matrix)

혼동행렬은 분류 모델의 예측 결과를 더 자세히 분석하기 위한 표입니다.

분류 모델이 무엇을 맞추고, 무엇을 틀렸는지 정리한 표

### 왜 필요한가
정확도만 보면 모델이 “얼마나 맞췄는지”는 알 수 있지만  
“어떻게 틀렸는지”는 잘 알 수 없습니다.  
혼동행렬은 이 부분을 자세히 보여줍니다.

### 4가지 구성 요소
Positive = 내가 찾아내고 싶은 쪽 / Negative = 그게 아닌 쪽

- TP(True Positive): 실제 Positive를 Positive로 맞힘
- TN(True Negative): 실제 Negative를 Negative로 맞힘
- FP(False Positive): 실제 Negative인데 Positive로 틀림
- FN(False Negative): 실제 Positive인데 Negative로 틀림

### 예시

질병 진단 문제에서 Positive = 환자 / Negative = 정상인

- TP: 실제 환자를 환자라고 맞힘
- TN: 실제 정상인을 정상이라고 맞힘
- FP: 실제 정상인을 환자라고 잘못 판단
- FN: 실제 환자를 정상이라고 잘못 판단

### 왜 중요한가
문제에 따라 더 위험한 실수가 다릅니다.
- 스팸 분류에서는 FP가 불편할 수 있음
- 질병 진단에서는 FN이 더 위험할 수 있음

### 핵심 이해 포인트
혼동행렬은 단순 정확도가 아닌 **오류의 종류까지 구분해 보여주는 표**입니다.

혼동행렬은 분류 모델의 예측 결과를 실제값과 비교하여, 맞춘 경우뿐 아니라 틀린 경우도 FP와 FN으로 나누어 어떤 종류의 오류가 발생했는지 보여주는 표이다.

### 헷갈리기 쉬운 포인트
FP와 FN의 차이를 꼭 구분해야 합니다.  
기준은 항상 **실제값과 예측값 비교**입니다.

## 1. RSS 뜻과 정의

RSS는 Residual Sum of Squares의 약자이며, 우리말로는 **잔차 제곱합**입니다.

RSS는 잔차제곱합이고, 최소제곱법은 그 RSS를 최소화하는 방법이다.

### 개념
잔차는 실제값과 예측값의 차이입니다.  
즉, 모델이 얼마나 틀렸는지를 나타내는 값입니다.

RSS는 이 잔차들을 제곱해서 모두 더한 값입니다.

### 공식
RSS = Σ(y - ŷ)²

- y: 실제값
- ŷ: 예측값

### 왜 제곱하는가
1. 오차가 음수든 양수든 모두 양수로 처리하기 위해  
2. 큰 오차에 더 큰 패널티를 주기 위해  
3. 수학적으로 미분과 최적화에 유리하기 위해

### 해석
- RSS가 작을수록 예측이 실제값에 가깝습니다.
- RSS가 크면 모델이 많이 틀렸다는 뜻입니다.

### 핵심 이해 포인트
RSS는 **회귀 모델의 전체 오차 크기**를 나타내는 값입니다.

### 헷갈리기 쉬운 포인트
RSS는 분류 지표가 아니라 **회귀 오차 관련 개념**입니다.

## 2. 신경망

신경망은 사람 뇌의 뉴런 연결 구조에서 영감을 받아 만든 모델입니다.

### 기본 구조
- 입력층(Input Layer)
- 은닉층(Hidden Layer)
- 출력층(Output Layer)

입력값은 여러 층을 거치며 계산되고, 최종 예측값이 출력됩니다.

### 기본 계산
각 노드는 보통 다음처럼 계산됩니다.

z = wx + b  
a = σ(z)

- w: 가중치
- b: 편향
- σ: 활성화 함수

즉, 입력을 단순히 더하는 것이 아니라  
가중치를 곱하고 편향을 더한 뒤, 비선형 함수를 적용합니다.

### 왜 중요한가
신경망은 단순 선형 모델보다 복잡한 패턴을 학습할 수 있습니다.  
특히 딥러닝은 이 신경망을 여러 층으로 깊게 쌓아 강력한 표현력을 얻습니다.

### 핵심 이해 포인트
신경망은 입력값에 가중치와 편향을 적용해 계산한 뒤, 활성화 함수를 거쳐 다음 층으로 전달하는 구조를 반복하여 학습하는 모델이다. 이를 통해 단순 선형 모델보다 복잡한 비선형 패턴을 학습할 수 있다.

신경망은 **입력 → 가중합 → 활성화 → 출력** 과정을 반복하며 학습합니다.

### 헷갈리기 쉬운 포인트
신경망은 단순히 “층이 많다”가 핵심이 아니라  
**비선형 관계를 학습할 수 있다**는 점이 중요합니다.

## 3. 경사하강법(Gradient Descent)

경사하강법은 손실함수를 줄이기 위해 파라미터를 반복적으로 업데이트하는 최적화 방법입니다.

### 직관적 이해
산 정상에서 가장 낮은 골짜기를 찾아 내려가는 상황을 생각하면 됩니다.  
현재 위치에서 가장 가파르게 내려가는 방향으로 조금씩 움직이면  
점점 더 낮은 곳으로 갈 수 있습니다.

이때
- 현재 위치: 현재 파라미터 값
- 높이: 손실값
- 내려가는 방향: 기울기의 반대 방향

### 업데이트 식
w ← w - α(dL/dw)

- w: 가중치
- α: 학습률
- dL/dw: 손실함수의 기울기

### 학습률의 의미
학습률은 한 번에 얼마나 이동할지를 정합니다.

- 너무 크면: 최솟값을 지나쳐 진동하거나 발산
- 너무 작으면: 학습이 매우 느림

### 왜 중요한가
딥러닝과 머신러닝에서 모델 학습은 결국  
손실을 줄이는 방향으로 파라미터를 조정하는 과정입니다.  
경사하강법은 그 핵심 방법입니다.

### 핵심 이해 포인트
경사하강법은 **손실함수를 최소화하기 위한 대표적인 최적화 알고리즘**입니다.

### 헷갈리기 쉬운 포인트
기울기 방향으로 가는 것이 아니라 **기울기의 반대 방향**으로 이동해야 손실이 줄어듭니다.

## 4. 로지스틱 회귀(Logistic Regression)

로지스틱 회귀는 이름과 달리 **분류 문제**, 특히 이진 분류에 많이 사용되는 모델입니다.

### 기본 아이디어
입력값을 선형 결합한 뒤 그 결과를 바로 사용하지 않고,  
시그모이드 함수에 넣어서 0과 1 사이의 값으로 변환합니다.

### 시그모이드 함수
σ(z) = 1 / (1 + e^(-z))

이 함수는 어떤 값이 들어와도 결과를 0과 1 사이로 바꿉니다.

### 왜 좋은가
출력을 확률처럼 해석할 수 있기 때문입니다.

예:
- 0.92 → Positive일 가능성이 높음
- 0.12 → Negative일 가능성이 높음

보통 0.5를 기준으로 클래스를 나눕니다.

### 어디에 쓰이는가
- 스팸 메일 분류
- 합격/불합격 예측
- 질병 유무 분류

### 핵심 이해 포인트
로지스틱 회귀는 **확률 기반 이진 분류 모델**입니다.

### 헷갈리기 쉬운 포인트
이름에 “회귀”가 들어가지만 실제 활용은 주로 **분류**입니다.

---

## 5. 손실함수(Loss Function)

손실함수는 모델이 얼마나 틀렸는지를 수치로 나타내는 함수입니다.

### 역할
모델 학습의 목표는 결국 손실값을 줄이는 것입니다.  
즉, 손실함수는 “지금 모델이 얼마나 못하고 있는가”를 수치로 알려줍니다.

### 예시
- 회귀: MSE, RSS
- 분류: Cross Entropy

### 왜 필요한가
모델은 스스로 잘하고 있는지 판단할 수 없습니다.  
그래서 손실함수가 현재 예측의 품질을 숫자로 알려주고,  
경사하강법은 그 손실을 줄이도록 파라미터를 조정합니다.

### 핵심 이해 포인트
손실함수는 **학습 방향을 정하는 기준**입니다.

### 헷갈리기 쉬운 포인트
손실함수와 평가 지표는 비슷해 보이지만 완전히 같지는 않습니다.  
손실함수는 주로 **학습 과정에서 최적화 대상**이고,  
평가 지표는 **결과 성능을 해석하기 위해** 사용합니다.


# 3차시

## 1. RNN 원리 이해

RNN(Recurrent Neural Network)은 순서가 있는 데이터를 처리하기 위한 신경망입니다.

### 왜 필요한가
문장, 음성, 시계열 데이터는 순서가 중요합니다.  
예를 들어 문장에서 단어 순서가 바뀌면 의미가 달라질 수 있습니다.

일반적인 모델은 입력 하나하나를 독립적으로 다루기 쉽지만,  
RNN은 이전 정보를 다음 계산에 반영하도록 설계되어 있습니다.

### 핵심 구조
- x_t: 현재 입력
- h_(t-1): 이전 시점 은닉 상태
- h_t: 현재 시점 은닉 상태

즉, 현재 계산에 과거 정보가 함께 반영됩니다.

### 장점
- 순서 정보 반영 가능
- 문장, 음성, 시계열 처리 가능

### 한계
시퀀스가 길어질수록 오래전 정보가 약해지는 문제가 있습니다.  
이를 **장기 의존성 문제**라고 합니다.

### 핵심 이해 포인트
RNN은 이전 상태를 기억하면서 현재 입력을 처리하는 순차 모델입니다.

### 헷갈리기 쉬운 포인트
RNN의 강점은 순차 처리지만, 긴 문맥 유지에는 약점이 있습니다.

---

## 2. NLP 성능평가

NLP: 자연어 처리(사람이 쓰는 언어를 컴퓨터가 다루는 분야)

- 이 문장이 긍정인지 부정인지 판단
- 문장을 번역
- 긴 글을 요약
- 다음 단어 예측

이런 것들이 다 NLP

NLP 성능평가는 작업 종류에 따라 달라집니다.

### 분류 문제 평가

정해진 여러 종류 중 하나를 맞히는 문제

결과가 정답 라벨로 나옴

예:
- 감성 분석
- 스팸 분류
- 문장 분류

자주 쓰는 지표:
- Accuracy: 전체 중에서 몇 개 맞췄는가
- Precision: Positive라고 예측한 것들 중, 진짜 Positive가 얼마나 되냐
- Recall: 실제 Positive인 것들 중, 내가 얼마나 찾아냈냐
- F1-score: Precision과 Recall을 함께 보는 균형 지표

### 생성 문제 평가

정답이 딱 하나의 라벨이 아니라, **문장을 직접 만들어내는 문제**

예:
- 번역
- 요약
- 문장 생성

자주 쓰는 지표:
- BLEU: 번역에서 자주 나옴, 모델이 만든 문장이 정답 문장과 단어/표현이 얼마나 겹치느냐
- ROUGE: 요약에서 많이 사용, 모델이 만든 요약이 정답 요약과 얼마나 겹치느냐
- Perplexity: 언어모델이 다음 단어를 얼마나 덜 헷갈려 하면서 예측하느냐

### 각 지표 느낌
- Accuracy: 전체 중 맞춘 비율
- Precision: Positive라고 예측한 것 중 실제 Positive 비율
- Recall: 실제 Positive 중 찾아낸 비율
- F1-score: Precision과 Recall의 균형 지표
- BLEU: 생성 문장이 정답 문장과 얼마나 겹치는지
- ROUGE: 요약 등에서 정답과의 중복 정도
- Perplexity: 언어모델이 다음 단어를 얼마나 잘 예측하는지

### 핵심 이해 포인트
NLP는 문제 종류에 따라 평가 방식이 다르며, 크게 보면 분류와 생성으로 나눔.

분류와 생성 문제의 지표를 구분할 수 있어야 합니다.

NLP는 문제마다 하는 일이 달라서 평가 방법도 다르다.
라벨 맞히기면 분류 지표를 쓰고, 문장 만들기면 생성 지표를 쓴다.

### 헷갈리기 쉬운 포인트
Accuracy는 만능 지표가 아닙니다.  
특히 생성형 NLP에서는 BLEU, ROUGE 같은 지표가 자주 등장합니다.

---

## 3. Transformer 원리 이해

Transformer는 현재 NLP와 거대 언어 모델의 핵심이 되는 구조입니다.

### 등장 배경
RNN은 순차적으로 처리하기 때문에 병렬화가 어렵고,  
긴 문맥을 다루는 데도 한계가 있었습니다.  
Transformer는 이를 개선하기 위해 등장했습니다.

### 핵심 개념: Self-Attention
문장 속 각 단어가 다른 단어들과 얼마나 관련 있는지를 계산합니다.

예를 들어
“나는 학교에 가서 공부했다”에서  
“공부했다”는 “학교”와 관계가 깊을 수 있습니다.

Self-Attention은 이런 관계를 계산해  
각 단어를 더 풍부한 문맥 정보와 함께 표현합니다.

### 장점
- 병렬 처리 가능
- 긴 문맥 관계 파악에 유리
- 대규모 학습에 적합

### 핵심 구성
- Self-Attention
- Positional Encoding
- Encoder / Decoder 구조

### 핵심 이해 포인트
Transformer는 **순차 처리 대신 attention 기반으로 문맥 전체 관계를 계산**합니다.

### 헷갈리기 쉬운 포인트
Transformer가 순서를 완전히 무시하는 것은 아닙니다.  
순서 정보는 **Positional Encoding**으로 보완합니다.

---

## 4. BERT 원리 이해

BERT는 Transformer의 **Encoder 기반** 사전학습 언어모델입니다.

### 특징
BERT의 큰 특징은 **양방향 문맥 이해**입니다.  
즉, 어떤 단어를 이해할 때 앞 단어뿐 아니라 뒤 단어도 함께 참고합니다.

예를 들어 “배”라는 단어가 나왔을 때  
주변 문맥을 함께 보며 과일인지 배(boat)인지 더 잘 구분할 수 있습니다.

### 대표 사전학습 방식
- Masked Language Modeling(MLM)
- Next Sentence Prediction(NSP)

#### MLM
문장 일부 단어를 가리고, 그 단어를 맞히게 학습합니다.

#### NSP
두 문장이 실제로 이어지는 문장인지 판단하게 학습합니다.

### 어디에 쓰이는가
- 감성 분석
- 문장 분류
- 질의응답
- 개체명 인식

### 핵심 이해 포인트
BERT는 Transformer Encoder 기반의 **양방향 언어모델**입니다.

### 헷갈리기 쉬운 포인트
BERT는 생성형 모델보다는 **이해 기반 NLP 작업**에서 강한 모델로 자주 설명됩니다.

---

## 5. LSTM 원리 이해

LSTM(Long Short-Term Memory)은 RNN의 장기 의존성 문제를 해결하기 위해 나온 구조입니다.

### 왜 필요한가
RNN은 문장이 길어질수록 앞 정보가 뒤까지 잘 전달되지 않는 문제가 있습니다.  
LSTM은 이 문제를 줄이기 위해 **정보를 선택적으로 저장/삭제**하는 구조를 갖습니다.

### 핵심 요소: 게이트 구조
- Forget Gate: 무엇을 버릴지 결정
- Input Gate: 무엇을 새로 저장할지 결정
- Output Gate: 무엇을 출력할지 결정

즉, 중요한 정보는 오래 유지하고  
불필요한 정보는 버릴 수 있습니다.

### 장점
- 긴 문맥 유지에 유리
- 순서 데이터 처리 성능 향상

### 핵심 이해 포인트
LSTM은 RNN보다 **긴 시퀀스의 정보를 더 잘 유지**할 수 있는 구조입니다.

### 헷갈리기 쉬운 포인트
LSTM은 RNN과 완전히 별개가 아니라  
**RNN의 한계를 보완한 발전형 구조**로 이해하는 것이 좋습니다.


# 4차시

## 1. 디코딩 알고리즘

언어모델은 매 시점마다 “다음에 어떤 토큰이 올 확률이 높은가”를 계산합니다.  
하지만 확률만 있다고 문장이 자동으로 완성되는 것은 아닙니다.  
이 확률분포에서 실제로 어떤 토큰을 선택할지 정하는 과정이 **디코딩**입니다.

### 왜 필요한가
같은 모델이라도 디코딩 방식에 따라
- 문장이 더 자연스러워질 수도 있고
- 더 단조로워질 수도 있고
- 더 창의적일 수도 있습니다.

즉, 디코딩은 생성 결과의 성격을 크게 바꿉니다.

### 대표 방식

#### Greedy Decoding
매 단계마다 확률이 가장 높은 토큰 하나만 선택합니다.
- 장점: 빠르고 단순함
- 단점: 전체적으로 더 나은 문장을 놓칠 수 있음

#### Beam Search
한 경로만 보지 않고 여러 후보를 유지하면서 탐색합니다.
- 장점: 더 좋은 전체 문장을 찾을 가능성 증가
- 단점: 계산량이 더 큼

#### Sampling
확률분포에 따라 무작위로 뽑습니다.
- 장점: 다양성 증가
- 단점: 불안정할 수 있음

#### Top-k Sampling
확률이 높은 상위 k개 후보만 남기고 그 안에서 샘플링합니다.

#### Top-p Sampling
누적 확률이 p가 될 때까지 후보를 남기고 그 안에서 샘플링합니다.

### 핵심 이해 포인트
디코딩은 **확률분포를 실제 문장 생성으로 바꾸는 선택 방식**입니다.

### 헷갈리기 쉬운 포인트
디코딩 알고리즘은 모델 자체가 아니라  
**모델 출력 사용 방식**에 가깝습니다.

---

## 2. 거대 언어 모델의 평가

거대 언어 모델(LLM)은 단순 정답률만으로 평가하기 어렵습니다.  
왜냐하면 언어 생성은 정답이 하나만 있는 문제가 아니고,  
정확성뿐 아니라 안전성, 유용성, 일관성도 중요하기 때문입니다.

### 평가 관점
- 정확성: 사실에 맞는가
- 추론 능력: 논리적으로 문제를 해결하는가
- 문맥 이해: 이전 대화나 문맥을 잘 반영하는가
- 안전성: 유해한 답변을 피하는가
- 일관성: 앞뒤 모순이 적은가
- 유용성: 사용자에게 실질적으로 도움이 되는가

### 평가 방식
- 벤치마크 데이터셋 기반 평가
- 사람 평가(Human Evaluation)
- 모델 기반 자동 평가

### 왜 중요한가
LLM은 단순 분류 모델이 아니라  
대화를 하고, 설명하고, 생성하는 모델이기 때문에  
평가 기준도 더 다면적이어야 합니다.

### 핵심 이해 포인트
LLM 평가는 **성능 + 안전성 + 사실성 + 유용성**을 함께 봅니다.

### 헷갈리기 쉬운 포인트
정확도 하나로 LLM 전체 성능을 판단하기는 어렵습니다.

---

## 3. 선호 학습(Preference Learning)

선호 학습은 사람의 선호를 반영해 모델 출력을 더 바람직하게 조정하는 학습 방식입니다.

### 왜 필요한가
언어모델이 단순히 다음 단어를 잘 맞힌다고 해서  
항상 사람이 원하는 답변을 하는 것은 아닙니다.

예를 들어
- 더 친절한 답변
- 더 안전한 답변
- 더 간결한 답변  
이 사람이 원하는 방향일 수 있습니다.

### 핵심 아이디어
여러 답변 후보가 있을 때  
사람이 “이 답변이 더 낫다”라고 평가한 데이터를 이용해  
모델이 선호되는 방향의 답변을 더 자주 내도록 학습합니다.

### 관련 키워드
- Human Feedback
- RLHF
- Preference Dataset

### 왜 중요한가
최근 생성형 AI는 단순 언어 예측을 넘어  
사람과 잘 상호작용하는 방향으로 발전하고 있습니다.  
선호 학습은 그 핵심 중 하나입니다.

### 핵심 이해 포인트
선호 학습은 **사람 기준에 더 맞는 응답을 하도록 모델을 조정하는 과정**입니다.

### 헷갈리기 쉬운 포인트
정답/오답만 학습하는 것이 아니라  
**여러 답변 중 어느 쪽을 더 선호하는가**를 반영한다는 점이 다릅니다.


# 5차시

## 1. CNN 원리 이해

CNN(Convolutional Neural Network)은 이미지 처리에 특히 강한 신경망 구조입니다.

### 왜 이미지에 적합한가
이미지는 픽셀들이 격자 형태로 배열되어 있고,  
가까운 픽셀끼리 의미 있는 패턴을 이루는 경우가 많습니다.

예:
- 선
- 모서리
- 질감
- 눈, 코, 입
- 물체 형태

CNN은 이런 **공간적 구조**를 잘 활용합니다.

### 핵심 아이디어
작은 필터(kernel)를 이미지 위에서 이동시키며  
지역적인 특징을 추출합니다.

완전연결층처럼 모든 픽셀을 한 번에 다 연결하지 않고,  
작은 영역을 반복적으로 보면서 특징을 뽑는다는 점이 핵심입니다.

### 장점
- 이미지 구조를 잘 활용함
- 파라미터 수를 줄일 수 있음
- 특징 추출에 강함

### 핵심 이해 포인트
CNN은 합성곱 연산을 통해 이미지의 **지역적 패턴과 공간적 특징**을 추출합니다.

### 헷갈리기 쉬운 포인트
CNN은 단순히 이미지용 모델이 아니라  
“공간적 구조가 중요한 데이터”에 강한 구조라고 볼 수 있습니다.

---

## 2. CNN 연산

CNN의 핵심 연산은 **합성곱(Convolution)** 입니다.

### 합성곱이란
작은 필터를 입력 위에 겹쳐 놓고  
각 위치에서 대응되는 값끼리 곱해 더한 뒤  
하나의 출력값을 만드는 연산입니다.

이 과정을 입력 전체에 반복하면  
새로운 특징맵(feature map)이 만들어집니다.

### 필터의 역할
필터마다 특정 패턴에 반응할 수 있습니다.

예:
- 가로선 검출
- 세로선 검출
- 가장자리 검출

즉, CNN은 여러 필터를 통해 다양한 특징을 뽑습니다.

### 왜 중요한가
이미지 분류에서 바로 “고양이”를 찾는 것이 아니라  
먼저 선, 가장자리, 질감 같은 저수준 특징을 찾고  
나중에 더 복잡한 특징으로 조합하는 구조를 가집니다.

### 핵심 이해 포인트
합성곱은 입력의 **지역 정보를 활용하여 특징을 뽑아내는 연산**입니다.

### 헷갈리기 쉬운 포인트
합성곱의 결과는 단순한 압축이 아니라  
“어떤 특징이 어디에 있는지”를 표현하는 특징맵입니다.

---

## 3. Pooling

Pooling은 특징맵의 크기를 줄이는 연산입니다.

### 왜 사용하는가
특징맵 크기가 너무 크면
- 연산량이 많아지고
- 파라미터가 늘어나고
- 과적합 가능성도 커집니다.

그래서 중요한 정보는 유지하면서  
공간 크기만 줄이는 과정이 필요합니다.

### 대표 종류
#### Max Pooling
작은 영역에서 가장 큰 값만 남깁니다.  
가장 강한 특징을 유지하는 느낌입니다.

#### Average Pooling
작은 영역의 평균값을 사용합니다.

### 효과
- 연산량 감소
- 과적합 완화
- 중요한 특징 강조
- 약간의 위치 변화에 덜 민감해짐

### 핵심 이해 포인트
Pooling은 **특징을 크게 훼손하지 않으면서 크기를 줄이는 과정**입니다.

### 헷갈리기 쉬운 포인트
Pooling은 필터로 특징을 추출하는 단계가 아니라  
이미 추출된 특징맵을 정리하는 단계에 가깝습니다.

---

## 4. 스트라이드 합성곱

Stride는 필터가 입력 위를 이동할 때 한 번에 몇 칸씩 움직이는지를 뜻합니다.

### 예
- Stride = 1 → 한 칸씩 이동
- Stride = 2 → 두 칸씩 이동

### 영향
Stride가 커질수록
- 출력 특징맵 크기는 작아지고
- 더 빠르게 계산되지만
- 세부 정보는 덜 보존될 수 있습니다.

즉, stride는 출력 크기와 정보량에 직접 영향을 줍니다.

### 왜 중요한가
CNN 설계에서는
- 얼마나 세밀하게 볼지
- 얼마나 다운샘플링할지  
를 stride로 조절할 수 있습니다.

### 핵심 이해 포인트
Stride는 **필터 이동 간격**이며, 출력 크기와 정보 손실 정도에 영향을 줍니다.

### 헷갈리기 쉬운 포인트
Stride는 필터 개수가 아니라 **이동 폭**입니다.


# 6차시

## 1. Vision-Language Models

Vision-Language Model(VLM)은 이미지와 텍스트를 함께 다루는 모델입니다.

### 왜 필요한가
현실의 많은 문제는 이미지나 글 중 하나만으로 해결되지 않습니다.

예:
- 사진을 보고 설명 생성하기
- 이미지에 대한 질문에 답하기
- 텍스트 설명과 맞는 이미지를 찾기

이런 작업은 시각 정보와 언어 정보를 함께 이해해야 합니다.

### 가능한 작업
- 이미지 캡셔닝
- 시각 질의응답(VQA)
- 이미지-텍스트 검색
- 멀티모달 이해

### 핵심 이해 포인트
VLM은 이미지와 텍스트를 같은 모델 안에서 연결해 해석하는 **멀티모달 모델**입니다.

### 헷갈리기 쉬운 포인트
이미지 모델과 언어 모델을 각각 따로 쓰는 것과 달리,  
VLM은 둘의 관계를 함께 학습한다는 점이 중요합니다.

---

## 2. SigLIP

SigLIP은 이미지와 텍스트의 대응 관계를 학습하는 비전-언어 정렬 모델 계열입니다.

### 큰 흐름에서 이해하기
이미지와 문장을 함께 다루는 모델에서는  
“이 이미지와 이 문장이 서로 잘 맞는가?”를 학습하는 것이 중요합니다.

SigLIP은 이런 정렬(alignment) 문제를 다루는 대표 개념 중 하나로 볼 수 있습니다.

### 특징
CLIP 계열처럼 이미지와 텍스트가 잘 맞는 쌍은 가깝게,  
잘 맞지 않는 쌍은 멀어지게 학습합니다.  
이 과정에서 sigmoid 기반 학습 방식을 사용하는 특징이 있습니다.

### 왜 중요한가
이미지 검색, 텍스트-이미지 매칭, 멀티모달 이해의 기반이 됩니다.

### 핵심 이해 포인트
SigLIP은 이미지와 텍스트 표현을 정렬하여  
둘의 의미적 연결을 학습하는 모델입니다.

### 헷갈리기 쉬운 포인트
시험에서는 세부 수식보다  
**이미지-텍스트 정렬 모델**이라는 큰 의미를 이해하면 좋습니다.

---

## 3. 이미지 생성 모델

이미지 생성 모델은 새로운 이미지를 만들어내는 모델입니다.

### 무엇을 하는가
예를 들어
- “우주복 입은 고양이”
- “비 오는 밤의 서울 거리”
같은 문장을 입력하면, 그 설명에 맞는 이미지를 생성할 수 있습니다.

### 대표 계열
- GAN
- Diffusion Model

### 핵심 원리
많은 이미지 데이터를 학습해서  
그 데이터가 가지는 분포를 이해한 뒤  
새로운 샘플을 만들어냅니다.

### 왜 중요한가
생성형 AI의 대표 분야 중 하나이며,
광고, 디자인, 콘텐츠 제작 등 다양한 분야와 연결됩니다.

### 핵심 이해 포인트
이미지 생성 모델은 **기존 데이터 분포를 학습해 새로운 이미지를 생성하는 모델**입니다.

### 헷갈리기 쉬운 포인트
이미지 분류 모델은 이미지를 보고 답을 맞히는 것이고,  
이미지 생성 모델은 **새로운 이미지를 만들어내는 것**입니다.

---

## 4. HuggingFace

HuggingFace는 다양한 AI 모델과 데이터셋, 라이브러리를 제공하는 대표적인 플랫폼/생태계입니다.

### 대표 기능
- 사전학습 모델 제공
- 데이터셋 제공
- 쉽게 모델 불러오기
- 파인튜닝 지원

### 대표 라이브러리
- transformers
- datasets
- diffusers

### 왜 중요한가
연구자나 개발자가 처음부터 모든 모델을 만들지 않아도  
이미 학습된 모델을 불러와 바로 실험하거나 응용할 수 있게 해줍니다.

### 핵심 이해 포인트
HuggingFace는 **AI 모델 활용과 실험을 쉽게 해주는 오픈 생태계**입니다.

### 헷갈리기 쉬운 포인트
단순 사이트가 아니라  
라이브러리, 모델 허브, 데이터셋 허브를 포함하는 생태계로 이해하는 것이 좋습니다.

---

## 5. AI 파운데이션 모델

파운데이션 모델은 대규모 데이터로 사전학습된 범용 기반 모델입니다.

### 왜 중요한가
예전에는 작업마다 모델을 따로 만드는 경우가 많았습니다.  
하지만 파운데이션 모델은 먼저 큰 범용 모델을 만들고,  
그 이후 여러 작업에 재활용하거나 미세조정하는 방식으로 활용됩니다.

### 특징
- 대규모 데이터 사전학습
- 다양한 작업에 전이 가능
- 텍스트, 이미지, 멀티모달 등 여러 형태 존재

### 예시 흐름
- 큰 언어모델 → 번역, 요약, 질의응답 등 여러 작업 활용
- 비전 모델 → 분류, 검출, 검색 등에 활용
- 멀티모달 모델 → 이미지+텍스트 작업에 활용

### 핵심 이해 포인트
파운데이션 모델은 **다양한 다운스트림 작업의 기반이 되는 대규모 사전학습 모델**입니다.

### 헷갈리기 쉬운 포인트
모든 AI 모델이 파운데이션 모델은 아닙니다.  
핵심은 **범용성, 대규모 사전학습, 재사용성**입니다.


# 자주 헷갈리는 비교 정리

## 1. AI / ML / DL
- AI: 가장 넓은 개념
- ML: 데이터로 규칙 학습
- DL: 신경망 기반 ML

## 2. Feature / Label
- Feature: 입력
- Label: 정답

## 3. 회귀 / 분류
- 회귀: 연속형 수치 예측
- 분류: 범주 예측

## 4. RSS / R²
- RSS: 오차의 크기
- R²: 설명력

## 5. RNN / LSTM / Transformer
- RNN: 순차 처리
- LSTM: 긴 문맥 보완
- Transformer: Attention 기반 전체 문맥 처리

## 6. CNN / Pooling / Stride
- CNN: 특징 추출 구조
- Pooling: 크기 축소
- Stride: 필터 이동 간격

## 7. 이미지 분류 모델 / 이미지 생성 모델
- 분류 모델: 이미지를 보고 답함
- 생성 모델: 새로운 이미지를 만듦

## 8. VLM / 파운데이션 모델
- VLM: 이미지와 텍스트를 함께 다루는 모델
- 파운데이션 모델: 여러 작업의 기반이 되는 범용 대규모 모델
