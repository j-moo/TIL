Django Form
- 사용자 입력 데이터를 수집하고, 처리 및 유효성 검사를 수행하기 위한 도구

> 유효성 검사를 단순화하고 자동화 할 수 있는 기능을 제공합니다.  
> 사용자가 잘못 입력한 데이터는 자동을 오류로 처리되어 안정성을 높입니다.  
> 개발자는 이를 통해 빠르고 일관된 입력 검증 기능을 구현할 수 있습니다.

**Django ModelForm (매우중요)**

Form vs ModelForm  

Form
- 사용자 입력 데이터를 DB에 저장하지 않을 때
(ex. 검색, 로그인)  

ModelForm
- 사용장 입력 데이터를 DB에 저장해야 할 때  
(ex. 게시글 작성, 회원가입)

ModelForm
- Model과 연결된 Form을 자동으로 생성해주는 기능을 제공

ModelForm class 정의
- 기존 ArticleForm 클래스 수정

Meta class
- ModelForm의 정보를 작성하는 곳  

> Meta class는 ModelForm 내부에서 어떤 모델과 연결할지, 어떤 필드를 사용할지 등을 정의하는 설정 공간입니다.
