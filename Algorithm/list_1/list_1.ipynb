알고리즘이란?

알고리즘(Algorithm)
- 문제를 해결하기 위한 절차나 방법입니다.

의사코드(슈도코드, Pseudocode)와 순서도
- 컴퓨터 분야에서 알고리즘을 표현하는 방법

```
CalSum(n)
    sum <- 0
    for i : 1 -> n
        sum <- sum + i
    return sum
```
