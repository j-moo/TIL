# DFS (Depth-First Search) 강의 노트

## 오늘의 한 줄
DFS는 **한 길로 최대한 깊게 들어갔다가**, 막히면 **되돌아와 다른 길을 찾는** 탐색 방법이다.  
(미로를 걷는 로봇을 떠올리면 이해가 빠르다.)

---

## 목표
- DFS가 무엇인지(왜 “깊이 우선”인지) 설명할 수 있다.
- `visited`가 왜 필요한지 알고, 빠짐없이 모든 정점을 방문할 수 있다.
- **재귀(시스템 스택)** 와 **직접 스택(반복문)** 두 방식으로 DFS를 구현할 수 있다.
- 탐색 순서를 바꾸고 싶을 때(오름차순/내림차순) 무엇을 조정해야 하는지 안다.


## 1) 핵심 개념

### DFS는 어떤 느낌이냐면…
- 길이 여러 갈래인 곳에서 **일단 한 갈래로 쭉 가본다.**
- 더 이상 갈 수 없으면(막다른 길) **마지막 갈림길로 되돌아가서** 다른 길을 간다.
- 이 “되돌아가기”를 가능하게 하는 도구가 **스택(stack)** 이다.  
  쉽게 말하면, “되돌아갈 주소(갈림길)를 적어두는 메모 더미” 같은 느낌이다.

---

### visited(방문 체크)가 꼭 필요한 이유
그래프(특히 사이클이 있는 그래프)는 **돌고 돌 수 있다.**  
그래서 “한 번 갔던 곳은 다시 안 간다”를 표시하지 않으면, DFS가 끝나지 않거나 같은 곳을 계속 방문할 수 있다.

- `visited[v] = True` : v를 이미 방문했음을 표시
- 다음에 v를 다시 만나면 **건너뛴다**

---

### DFS vs BFS (감각만 잡기)
- DFS: 한 길로 **깊게** (스택/재귀)
- BFS: 가까운 곳부터 **넓게** (큐)

| 구분 | 핵심 동작 | 자주 쓰는 도구 | 잘 어울리는 상황 |
|---|---|---|---|
| DFS | 깊게 파고들기 → 막히면 되돌아오기 | 스택 / 재귀 | 경로 탐색, 백트래킹, 연결요소, 사이클 확인 |
| BFS | 가까운 곳부터 차근차근 | 큐 | 최단거리(간선 가중치 없을 때), 레벨 탐색 |


## 2) 대표 예시(그림 따라가기)

아래는 “정점 A에서 시작해서, 인접 정점은 **오름차순**으로 고른다”는 규칙으로 DFS를 수행하는 예시다.  
(원본 강의 흐름을 살리면서, 문장만 조금 더 매끄럽게 정리했어.)

### 예시 진행(원본 내용 기반)
초기 상태: `visited`를 모두 `False`로 만들고, 공백 스택을 준비한다.

1. 정점 **A**에서 DFS 시작  
2. A의 미방문 인접 정점 **B, C**가 있다 → A를 스택에 push, 오름차순으로 **B** 선택  
3. B의 미방문 인접 정점 **D, E**가 있다 → B를 스택에 push, 오름차순으로 **D** 선택  
4. D의 미방문 인접 정점 **F**가 있다 → D를 스택에 push, **F** 선택  
5. F의 미방문 인접 정점 **E, G**가 있다 → F를 스택에 push, 오름차순으로 **E** 선택  
6. E의 미방문 인접 정점 **C**가 있다 → E를 스택에 push, **C** 선택  
7. C는 더 갈 곳이 없다 → 스택 pop 해서 **E**로 돌아가 확인  
8. E도 더 갈 곳이 없다 → pop 해서 **F**로 돌아가 확인  
9. F에는 미방문 인접 정점 **G**가 있다 → F를 push, **G** 선택  
10. G는 더 갈 곳이 없다 → pop 해서 **F** 확인  
11. F도 더 갈 곳이 없다 → pop 해서 **D** 확인  
12. D도 더 갈 곳이 없다 → pop 해서 **B** 확인  
13. B도 더 갈 곳이 없다 → pop 해서 **A** 확인  
14. A도 더 갈 곳이 없고, 스택도 비었다 → DFS 종료

## 3) 풀이 전략(생각 순서)

DFS를 구현하는 방법은 크게 두 가지다.

### A. 재귀 DFS (함수가 스택을 대신 써줌)
- “지금 정점 방문 → 이웃 중 안 간 곳으로 들어가기”를 함수로 만든다.
- 함수 호출이 쌓이는 게 곧 **시스템 스택**이다.

### B. 스택 DFS (반복문으로 직접 구현)
- 우리가 `stack` 리스트를 직접 만들어서
- `pop()` 하면서 탐색을 진행한다.

---

### 탐색 순서(오름차순)를 맞추는 팁
재귀는 보통 `for next in sorted(adj[v])`로 해결된다.

반복문(스택) 방식은 **push 순서가 결과에 영향을 준다.**  
스택은 LIFO라서, 오름차순으로 방문하고 싶으면 보통 이웃을 **역순으로 push**한다.

예)
- 방문을 `A -> B -> C` 오름차순으로 하고 싶다
- 스택에는 `C`를 먼저 넣고, `B`를 나중에 넣어야 `B`가 먼저 pop 된다.


## 4) 의사코드(Pseudocode)

### 재귀 DFS
```text
DFS(v):
    visited[v] = True
    visit(v)          # 출력/기록 등
    for each nxt in adj[v] (정렬 가능):
        if not visited[nxt]:
            DFS(nxt)
```

### 스택 DFS (반복)
```text
stack = [start]
while stack not empty:
    v = stack.pop()
    if visited[v]: 
        continue
    visited[v] = True
    visit(v)

    for each nxt in adj[v] (역순으로 push하면 오름차순 방문 가능):
        if not visited[nxt]:
            stack.push(nxt)
```


````python
# 그래프를 "인접 리스트"로 표현해보자.
# (이미지 예시와 비슷하게 A~G를 사용)

from collections import defaultdict

graph = {
    "A": ["B", "C"],
    "B": ["A", "D", "E"],
    "C": ["A", "E"],
    "D": ["B", "F"],
    "E": ["B", "C", "F"],
    "F": ["D", "E", "G"],
    "G": ["F"],
}

# 탐색 순서를 통일하기 위해 인접 리스트를 오름차순 정렬해둔다.
for v in graph:
    graph[v] = sorted(graph[v])
````

````python
def dfs_recursive(start: str, graph: dict[str, list[str]]) -> list[str]:
    """재귀 DFS: 방문 순서를 리스트로 반환"""
    visited = set()
    order: list[str] = []

    def _dfs(v: str):
        visited.add(v)
        order.append(v)
        for nxt in graph.get(v, []):
            if nxt not in visited:
                _dfs(nxt)

    _dfs(start)
    return order
````

````python
def dfs_stack(start: str, graph: dict[str, list[str]]) -> list[str]:
    """스택 DFS(반복): 방문 순서를 리스트로 반환

    - 오름차순 방문을 원하면, 스택에는 이웃을 역순으로 push한다.
    """
    visited = set()
    order: list[str] = []
    stack = [start]

    while stack:
        v = stack.pop()
        if v in visited:
            continue
        visited.add(v)
        order.append(v)

        # 오름차순 방문을 위해 역순 push
        for nxt in reversed(graph.get(v, [])):
            if nxt not in visited:
                stack.append(nxt)

    return order
````

````python
print("재귀 DFS:", dfs_recursive("A", graph))
print("스택 DFS:", dfs_stack("A", graph))
````

## 5) 테스트 + 자주 하는 실수(엣지 케이스)

### 실수 1) visited를 너무 늦게 처리해서 중복 push가 많이 생김
- “pop 했을 때 방문 체크” 방식은 안전하지만, 같은 정점이 스택에 여러 번 들어갈 수 있다.
- 큰 그래프에서는 비효율이 될 수 있어서, 상황에 따라 **push할 때 visited 처리**를 하기도 한다.

### 실수 2) 스택 DFS에서 순서가 뒤집힘
- `for nxt in graph[v]: stack.append(nxt)` 그대로 넣으면,
  스택 특성(LIFO) 때문에 실제 방문 순서가 기대와 다를 수 있다.
- 오름차순 방문이 목적이면 **역순으로 push**하는 패턴을 자주 쓴다.

### 실수 3) 그래프가 끊어져(Disconnected) 있는 경우
- 시작 정점에서 닿을 수 있는 곳만 방문하고 끝난다.
- “그래프 전체를 빠짐없이” 방문하려면, 모든 정점에 대해 DFS를 시작하는 루프가 필요하다.

아래는 “그래프 전체 방문(연결요소까지)” 예시 코드다.


````python
def dfs_all_nodes(graph: dict[str, list[str]]) -> list[list[str]]:
    """그래프가 여러 덩어리(연결요소)로 나뉘어 있어도 전체를 방문한다.
    반환: 각 연결요소의 방문 순서 리스트들의 모음
    """
    visited = set()
    components: list[list[str]] = []

    for start in sorted(graph.keys()):
        if start in visited:
            continue

        order: list[str] = []
        stack = [start]
        while stack:
            v = stack.pop()
            if v in visited:
                continue
            visited.add(v)
            order.append(v)
            for nxt in reversed(graph.get(v, [])):
                if nxt not in visited:
                    stack.append(nxt)

        components.append(order)

    return components

print(dfs_all_nodes(graph))
````

## 6) 시간/공간 복잡도

- 시간 복잡도: **O(V + E)**
  - 각 정점(V)은 최대 한 번 “방문 처리”되고,
  - 각 간선(E)은 인접 리스트를 훑을 때 한 번(또는 두 번) 확인된다.

- 공간 복잡도: **O(V)**
  - `visited`에 정점을 저장해야 하고
  - 재귀면 호출 스택, 반복이면 직접 만든 스택이 최대 V까지 쌓일 수 있다.

> 참고: 파이썬 재귀 DFS는 깊이가 너무 깊으면 재귀 제한에 걸릴 수 있다.  
> 그럴 땐 스택(반복) 방식이 더 안전하다.


## 7) 요약 + 미니 퀴즈

### 요약(3줄)
- DFS는 **한 방향으로 끝까지** 갔다가, 막히면 **되돌아오는** 탐색이다.
- 되돌아오려면 “갈림길 기록”이 필요하고, 그 역할을 **스택**(또는 재귀 호출 스택)이 한다.
- 그래프에서는 **visited가 필수**다(사이클 때문에).
