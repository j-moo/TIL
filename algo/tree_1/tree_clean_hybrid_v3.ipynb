## 1. 핵심 개념

### 트리란?
- **비선형 자료구조**: 한 줄로 쭉 이어진 게 아니라, 가지가 뻗는 구조
- **계층 구조**: 위(상위)에서 아래(하위)로 내려가며 연결

쉽게 말하면, **가족 관계도(족보)** 처럼 생각하면 편해요.  
- 가장 위에 “조상”이 있고(루트), 그 아래로 자식들이 있고, 또 그 아래로 손자들이 있어요.

### 용어 정리(자주 나오는 것만 먼저)
- **노드(node)**: 트리의 각 점(원소)
- **간선(edge)**: 노드를 잇는 선
- **루트(root)**: 맨 위 시작 노드
- **부모/자식(parent/child)**: 위-아래로 연결된 관계
- **형제(sibling)**: 같은 부모를 가진 노드들
- **리프(leaf, 단말)**: 자식이 없는 끝 노드
- **서브트리(subtree)**: 어떤 노드를 뿌리로 하는 “부분 트리”

## 2. 이진 트리(Binary Tree)

### 정의
- 각 노드가 가질 수 있는 자식이 **최대 2개**인 트리
- 보통 **왼쪽 자식 / 오른쪽 자식**을 구분해서 저장

![이진 트리 예시](<화면 캡처 2026-03-02 234948.jpg>)

### 특성(자주 쓰는 공식)
- 레벨 `i`에서 노드의 최대 개수: `2**i`
- 높이가 `h`인 이진 트리의 노드 개수
  - 최소: `h + 1` (한 줄로만 내려가는 경우)
  - 최대: `2**(h + 1) - 1` (포화 이진 트리)

![이진 트리 특성](<화면 캡처 2026-03-02 235119.jpg>)

### 종류(그림이랑 같이 기억하면 쉬움)
- **포화 이진 트리(Full)**: 모든 레벨이 꽉 찬 형태
- **완전 이진 트리(Complete)**: 마지막 레벨만 “왼쪽부터” 차는 형태
- **편향 이진 트리(Skewed)**: 한쪽으로만 계속 내려가는 형태(막대기처럼)

## 3. 트리 표현(코드로 저장하는 방법)

### (1) 완전/포화 이진 트리라면: **배열 인덱스 규칙**이 강력함
- 루트를 1번 인덱스로 두면 규칙이 아주 깔끔해요.
- **부모**: `i // 2`
- **왼쪽 자식**: `2 * i`
- **오른쪽 자식**: `2 * i + 1`
- 레벨 `n`의 시작 번호: `2**n`

![배열로 표현(1)](<화면 캡처 2026-03-03 000009.jpg>)
![노드 번호 규칙](<화면 캡처 2026-03-03 000155.jpg>)
![레벨과 번호](<화면 캡처 2026-03-03 000208.jpg>)

배열 크기는 높이 `h` 기준으로 보통 `2**(h+1)` 정도를 잡습니다.  
(쉽게 말하면: “마지막 레벨까지 들어갈 자리”를 미리 통째로 확보하는 느낌)

다만, **편향 트리**처럼 빈 칸이 많은 경우에는 배열이 “빈 자리까지” 다 챙기게 돼서 메모리가 아까워요.

![배열 표현의 단점(편향)](<화면 캡처 2026-03-03 000648.jpg>)

---

### 코드로 보기: 배열 인덱스 규칙을 실제로 쓰는 예제
- 아래 코드는 **완전 이진 트리**를 `tree[1..N]` 배열로 두고,
  **중위 순회**를 하면서 `1,2,3,...`을 차례대로 채워 넣습니다.
- 온라인 저지 입력이 필요해서, 셀을 그대로 실행하려면 문제 입력이 있어야 해요.
  (입력 없이 확인하고 싶으면, 뒤쪽의 ‘입력 없는 미니 예제’ 셀을 먼저 참고하면 좋아요.)


````python
# 실습_이진 탐색

T = int(input())

for tc in range(1, T + 1):
    # 노드의 개수가 N개인 완전 이진 트리
    N = int(input())
    # 노드의 개수가 N개인 비어있는 트리 만들기(노드번호 1부터 시작)
    tree = [0] * (N + 1)

    # +1 할 숫자(노드 안에 적을 숫자)
    num = 1


    # 이 비어있는 트리를 중위순회 하면서
    # 만나는 노드에 1씩 증가하는 숫자를 적으면 끝
    def inorder(t):
        global num
        if t <= N:  # t가 존재하는 노드면:
            # t번 노드의 왼쪽 서브 트리순회
            inorder(t * 2)
            # t번 노드에 +1 한 숫자 쓰기
            tree[t] = num
            # print(t, num)
            num += 1
            # t번 노드의 오른쪽 서브 트리순회
            inorder(t * 2 + 1)


    # 1번 노드에서 중위순회 시작
    inorder(1)
    print(f"#{tc} {tree[1]} {tree[N // 2]}")
````

### 코드 해설(왜 이렇게 동작할까?)

- `tree = [0] * (N + 1)`  
  노드 번호를 **그대로 인덱스**로 쓰기 위해 0번 칸을 비워둡니다. (`1`이 루트)

- 자식으로 내려갈 때 `t*2`, `t*2+1`을 쓰는 이유  
  완전 이진 트리를 배열로 저장하면,
  - 왼쪽 자식 = `2*t`
  - 오른쪽 자식 = `2*t + 1`
  규칙이 항상 성립해요.

- `if t <= N:` 체크  
  “존재하지 않는 노드 번호”까지 내려가면 멈추게 하는 **경계 조건**입니다.  
  쉽게 말하면: *트리 바깥으로 나가면 돌아오게 하는 안전장치*예요.

- 중위 순회(inorder)에서 `tree[t] = num`이 가운데에 있는 이유  
  중위는 `왼쪽 → 나 → 오른쪽`이라서,  
  왼쪽을 다 채운 다음에 현재 노드에 값을 쓰고, 그 다음 오른쪽을 채우는 흐름이 됩니다.

- 출력 `tree[1]`과 `tree[N//2]`  
  문제에서 보통 “루트 값”과 “N//2(중간 노드) 값”을 묻는 패턴이 자주 나옵니다.

### 코드로 보기: `cleft/cright` (부모 → 자식 저장)
- `cleft[p]` : p번 노드의 **왼쪽 자식**
- `cright[p]` : p번 노드의 **오른쪽 자식**

이 방식은 입력이 `(부모, 자식)` 쌍으로 들어올 때 제일 많이 씁니다.


````python
# 트리 만들기 1.py

# 마지막 노드 V, 간선의 개수 N
V, N = map(int, input().split())
# 트리의 정보가 한줄 입력으로 들어온다.
tree = list(map(int, input().split()))
# 5 4
# 1 2 1 3 3 4 3 5
# 1-2/1-3/3-4/3-5/

# 부모 노드 번호를 인덱스로 저장하는 방법
cleft = [0] * (V + 1)
cright = [0] * (V + 1)
# cleft[4] = 4번 노드의 왼쪽 자식 번호
# cright[1] = 1번 노드의 오른쪽 자식 번호

# 한줄 입력을 간선 개수만큼 자른다.
for i in range(N):
    # 부모 번호
    p = tree[2 * i]
    # 자식 번호
    c = tree[2 * i + 1]

    # p의 자식은 c번이다.
    # 이진트리에서 자식은 왼쪽? 오른쪽?

    # 먼저 왼쪽 자식이 있나 확인
    # p번노드의 왼쪽 자식이 없다면
    if cleft[p] == 0:
        # 왼쪽 자식으로 c 넣기
        cleft[p] = c
    # p번 노드의 왼쪽 자식이 있었다면
    else:
        # 오른쪽 자식으로 c 넣기
        cright[p] = c

print(cleft)
print(cright)
````

### 코드 해설

- `V, N = map(int, input().split())`  
  `V`는 마지막 노드 번호(= 노드 개수), `N`은 간선 개수예요.  
  (문제마다 이름이 달라질 수 있지만, 여기서는 주석대로 이해하면 됩니다.)

- `tree = list(map(int, input().split()))`  
  한 줄로 `부모 자식 부모 자식 ...` 형태로 주어지는 입력을 통째로 받습니다.

- `cleft`, `cright`를 0으로 초기화  
  0은 “아직 자식이 없다”는 뜻의 **비어있음 표시**로 씁니다.

- 간선을 하나씩 보면서
  `cleft[p]`가 비어 있으면 왼쪽에 넣고, 아니면 오른쪽에 넣습니다.  
  쉽게 말하면: *한 부모에게 자식이 들어오면 왼쪽부터 차곡차곡 채우는 느낌*이에요.

- 이렇게 만들면 `preorder/inorder/postorder` 같은 순회에서
  `cleft[t]`, `cright[t]`로 다음 노드를 바로 찾을 수 있어요.

### 코드로 보기: `parent` (자식 → 부모 저장)
- `parent[c] = p` 형태로 저장해두면,
  어떤 노드 `x`의 부모를 `parent[x]`로 바로 찾을 수 있어요.

이 방식은 특히 **루트 찾기 / 조상 따라 올라가기**에 강합니다.


````python
# 트리 만들기 2.py

# 마지막 노드 V, 간선의 개수 N
V, N = map(int, input().split())
# 트리의 정보가 한줄 입력으로 들어온다.
tree = list(map(int, input().split()))

# 자식 번호를 인덱스로 부모 번호를 저장
parent = [0] * (V + 1)
# parent[2] = 2번자식의 부모 노드 번호
# parent[4] = 4번자식의 부모 노드 번호
# x번 노드의 부모 노드 번호를 알고싶어요 = parent[x]

# 한줄 입력을 간선 개수만큼 자른다.
for i in range(N):
    # 부모 번호
    p = tree[2 * i]
    # 자식 번호
    c = tree[2 * i + 1]

    # c번 노드의 부모 노드 번호는 p번이다.
    parent[c] = p

print(parent)

# 5번 노드의 조상 노드 모두 찾고 싶다.
child = 5

ancestor = []

while parent[child] != 0:
    # child 의 부모 노드가 0이 아니다 => child는 루트노드가 아니다.
    # child 의 부모 위로 한칸 올라가 본다.
    child = parent[child]
    # 이 노드는 부모노드니까 조상목록에 추가
    ancestor.append(child)

# 어느 순간 child 의 부모 번호가 0번이 된다. == child는 루트 노드다.
root = child
print(root, ancestor)
````

### 코드 해설

- `parent = [0] * (V + 1)`  
  부모가 없는 노드는 `0`으로 남습니다. (보통 그게 **루트**)

- 입력을 `(p, c)`로 보면서 `parent[c] = p`로 채웁니다.  
  즉, “자식 칸에 부모를 써놓는 메모”라고 생각하면 돼요.

- 루트를 찾고 싶다면?  
  모든 노드를 보면서 `parent[x] == 0`인 노드를 찾으면 됩니다.  
  (단, 문제에서 루트가 여러 개(포레스트)일 수도 있으면 예외 처리가 필요)

- 조상을 찾고 싶다면?  
  `x = parent[x]`를 반복하면 위로 계속 올라갑니다.  
  쉽게 말하면: *가족 관계도에서 ‘부모 → 부모 → 부모’로 위로 타고 올라가는 것*이에요.

### 코드로 보기: Node 클래스로 연결해서 트리 만들기
배열로 저장하기 어려운 경우(노드 번호가 복잡하거나, 노드를 “필요할 때만” 만들고 싶을 때)는
`Node` 객체를 만들어 `left/right`에 직접 연결하는 방식도 많이 씁니다.


````python
# 트리 만들기 3.py

class Node:
    def __init__(self, value):
        self.left = None
        self.right = None
        self.value = value


A = Node("A")
B = Node("B")
C = Node("C")
D = Node("D")
E = Node("E")

A.left = D
A.right = E
E.left = B
E.right = C
# 전위순회 : V - L - R
def preorder(t):
    # t번 노드가 존재하는지 확인
    if t:
        # V 노드에서 할거
        print(t.value, end=" ")
        # L 순회 t노드의 왼쪽 전위순회
        preorder(t.left)
        # R 순회 t노드의 오른쪽 전위순회
        preorder(t.right)

preorder(A)
````

### 코드 해설

- `Node`는 노드 1개를 표현하는 ‘상자’예요.
  - `value`: 노드에 들어있는 값
  - `left`, `right`: 왼쪽/오른쪽 자식 노드를 가리키는 화살표(참조)

- `A.left = D`처럼 연결하는 부분이 “간선(edge)을 만든다”는 뜻입니다.

- 마지막의 `preorder(A)`는 **전위 순회(VLR)** 예시예요.
  `print → left → right` 순서로 내려가며 방문합니다.

- 이 방식의 장점
  - 노드 번호에 얽매이지 않음
  - 필요한 노드만 만들어서 연결 가능

- 단점
  - 입력으로 들어오는 `(부모, 자식)`을 바로 연결하려면
    노드 객체를 미리 딕셔너리로 관리하는 등의 작업이 필요할 수 있어요.


## 4. 순회(Traversal)

트리 순회는 “노드를 한 번씩 전부 방문”하는 방법이에요.  
쉽게 말하면, **미로를 탐험하듯이** “어떤 순서로 들를지” 정하는 규칙입니다.

### 3가지 기본 순회
- **전위(Preorder) = VLR**: (나 방문) → (왼쪽) → (오른쪽)
- **중위(Inorder) = LVR**: (왼쪽) → (나 방문) → (오른쪽)
- **후위(Postorder) = LRV**: (왼쪽) → (오른쪽) → (나 방문)

### 코드로 보기: 전/중/후위 순회(배열 + 재귀)
아래 코드는 입력으로 간선을 받아 `cleft/cright`를 만든 뒤,
전위/중위/후위 순회를 그대로 구현한 예시입니다.


````python
# 트리 순회.py

# 13
# 1 2 1 3 2 4 3 5 3 6 4 7 5 8 5 9 6 10 6 11 7 12 11 13

# 트리 노드 수
N = int(input())
# 트리 정보 한줄
tree = list(map(int, input().split()))
# 한줄에서 두개씩 잘라서 부모 / 자식 노드 트리 만들기

cleft = [0] * (N + 1)
cright = [0] * (N + 1)

# 두개씩 쪼개야 하는 횟수는 N-1
for i in range(N - 1):
    # 앞이 부모 번호
    p = tree[i * 2]
    # 뒤가 자식 번호
    c = tree[i * 2 + 1]

    # p번 노드의 왼쪽 자식있는지 체크
    if cleft[p] == 0:
        cleft[p] = c
    else:
        cright[p] = c

print(cleft)
print(cright)


# 1. 전위순회
def preorder(t):
    # t번 노드가 존재하면
    if t:
        # t번 노드에서 처리해야 할일
        print(t, end=" ")
        # t번 노드 왼쪽
        preorder(cleft[t])
        # t번 노드 오른쪽
        preorder(cright[t])

# 2. 중위순회
def inorder(t):
    # t번 노드가 존재하면
    if t:
        # t번 노드 왼쪽
        inorder(cleft[t])
        # t번 노드에서 처리해야 할일
        print(t, end=" ")
        # t번 노드 오른쪽
        inorder(cright[t])

# 3. 후위순회
def postorder(t):
    # t번 노드가 존재하면
    if t:
        # t번 노드 왼쪽
        postorder(cleft[t])
        # t번 노드 오른쪽
        postorder(cright[t])
        # t번 노드에서 처리해야 할일
        print(t, end=" ")

preorder(1)
print()
inorder(1)
print()
postorder(1)
print()

# 1 2 4 7 12 3 5 8 9 6 10 11 13
# 12 7 4 2 1 8 5 9 3 10 6 13 11
# 12 7 4 2 8 9 5 10 13 11 6 3 1
````

### 코드 해설

- 먼저 `cleft/cright`를 만드는 이유  
  순회는 결국 “현재 노드에서 왼쪽/오른쪽으로 어떻게 이동하나?”가 핵심이라서,
  이동 표지판(`cleft/cright`)을 먼저 만들어두는 겁니다.

- `if t:` 조건  
  `t == 0`이면 노드가 없다는 뜻이므로 더 내려가지 않습니다.

- 전위/중위/후위의 차이는 **print 위치** 딱 하나입니다.
  - 전위(VLR): `print`가 맨 위
  - 중위(LVR): `print`가 가운데
  - 후위(LRV): `print`가 맨 아래

> 실행 주의: 이 셀은 `input()`을 받기 때문에, 문제 입력이 없으면 실행이 멈춥니다.

### 대표 예시: 중위 순회로 문자열 만들기(문제형)
- 각 노드에 알파벳이 들어 있고, 중위 순회로 방문한 순서대로 이어 붙이면 정답이 되는 유형입니다.
- 입력 줄 길이에 따라(2/3/4개) 자식이 없는 리프/자식 1개/자식 2개를 구분하는 게 포인트예요.


````python
# 실습_중위순회

T = 10

for tc in range(1, T + 1):
    # 노드의 개수
    N = int(input())

    # 왼쪽 자식
    cleft = [0] * (N + 1)
    # 오른쪽 자식
    cright = [0] * (N + 1)

    # 어떤 인덱스에 어떤 알파벳이 저장되어있는지 확인할 리스트
    node = [0] * (N + 1)
    # node[1] => 1번 노드에 저장된 알파벳은 무엇인가?
    # node[1] == 'A' : 1번 노드에 저장된 알파벳은 'A'

    for i in range(N):
        # 한 줄에 노드 하나의 정보가 들어온다.
        # 노드번호, 알파벳, 왼쪽자식, 오른쪽 자식
        # 한줄 입력을 다 받아놓고 쪼개서 개수를 센다음에 개수에 맞게 처리
        info = input().split()
        # info 의 길이가 2 : 단말노드(잎 노드), 자식 노드가 없음
        # info 의 길이가 3 : 자식노드가 1개 있음(왼쪽 자식)
        # info 의 길이가 4 : 자식노드가 2개 있음(왼쪽, 오른쪽)
        p = int(info[0])  # 노드 번호
        alpha = info[1]  # 해당 노드의 알파벳
        node[p] = alpha

        # 길이가 3 이상인 경우(자식 노드가 있는 경우)
        if len(info) >= 3:
            # 왼쪽 자식
            cleft[p] = int(info[2])
            # 길이가 4면
            if len(info) == 4:
                # 오른쪽 자식도 있음
                cright[p] = int(info[3])

    # 문제에서 원하는 답 구하기
    answer = ""

    # 중위순회
    def inorder(t):
        global answer
        # t번 노드가 존재한다면
        if t:
            # t번 노드의 왼쪽 서브트리 탐색 : L
            inorder(cleft[t])
            # t번 노드 방문 처리 : V
            # t번노드에 써져있는 알파벳을 이어붙이기
            answer += node[t]
            # t번 노드의 오른쪽 서브트리 탐색 : R
            inorder(cright[t])

    inorder(1)
    # 순회가 끝나면 단어 완성
    print(f"#{tc} {answer}")
````

### 코드 해설

- `info = input().split()` 이후 `len(info)`로 자식 존재 여부를 판단합니다.
  - 길이 2: 리프(자식 없음)
  - 길이 3: 왼쪽 자식만 있음
  - 길이 4: 왼쪽/오른쪽 자식 모두 있음

- `answer += node[t]`가 중위 순회의 “방문(V)” 단계에 들어가 있는 이유  
  중위는 `왼쪽 → 나 → 오른쪽`이므로, 방문 순간에 문자열을 이어 붙이면 됩니다.

> 실행 주의: 이 셀도 문제 입력이 필요합니다. (SWEA 1231 같은 형식)

### 참고: 완성 예시(입력 없이도 실행 가능)
아래는 같은 원리를 **끝까지 완성한 형태**이고, 미니 테스트 트리도 같이 넣어둔 셀입니다.


````python
# 추가 참고용: 수식 트리 평가(완성 예시)
# 입력 형식은 SWEA 1232(사칙연산) 스타일을 가정합니다.
# - 노드 i: "i 값 [왼쪽] [오른쪽]"
# - 값이 숫자면 리프, 값이 연산자면 내부 노드

def eval_expr(t):
    if node[t].isdigit():
        return float(node[t])
    op = node[t]
    left = eval_expr(cleft[t])
    right = eval_expr(cright[t])
    if op == '+':
        return left + right
    elif op == '-':
        return left - right
    elif op == '*':
        return left * right
    elif op == '/':
        return left / right

# 미니 테스트(입력 없이)
#    *
#   / \
#  +   4
# / \
# 1  2
N = 5
node = [''] * (N+1)
cleft = [0]*(N+1)
cright = [0]*(N+1)

node[1] = '*'; cleft[1]=2; cright[1]=5
node[2] = '+'; cleft[2]=3; cright[2]=4
node[3] = '1'
node[4] = '2'
node[5] = '4'

print("식 결과:", int(eval_expr(1)))  # (1+2)*4 = 12
````

### 코드 해설(완성 예시)

- `eval_expr(t)`는 “t를 루트로 하는 서브트리의 값”을 반환합니다.
- 내부 노드면 재귀로 `left`, `right`를 먼저 구한 뒤 연산자를 적용합니다.
- 맨 마지막 `eval_expr(1)`은 루트(1번 노드)에서 계산을 시작한다는 뜻입니다.


## 6. 풀이 전략(문제에서 자주 쓰는 패턴)

### 6-1) 간선 입력으로 트리 만들기
- 입력이 `(부모, 자식)` 쌍으로 주어지면 보통 `cleft/cright` 혹은 `parent`를 만듭니다.
- 이진 트리라면, 한 부모는 자식을 최대 2개만 가지므로  
  `cleft[p]`가 비어 있으면 왼쪽에 넣고, 아니면 오른쪽에 넣는 패턴이 흔해요.

### 6-2) 순회는 재귀가 가장 직관적
- 전위: `visit(t) -> left -> right`
- 중위: `left -> visit(t) -> right`
- 후위: `left -> right -> visit(t)`

### 6-3) 수식 트리 계산은 “후위 순회”
- 리프(숫자)는 그대로 값
- 내부 노드(연산자)는 왼쪽값/오른쪽값을 받아 계산

> 포인트: 트리 문제는 **(1) 표현 방식 선택**이 절반을 먹고 들어갑니다.


---

### 대표 패턴: 서브트리 노드 개수 세기
서브트리 크기는 사실 ‘그 노드를 루트로 전위/후위/중위 아무거나 한 번 돌면서 방문 수를 세면’ 끝입니다.


````python
# subtree.py
T = int(input())

for tc in range(1, T + 1):
    # 간선의개수 E개 (노드의 개수는 E+1)
    # N은 시작 노드 번호
    E, N = map(int, input().split())

    cleft = [0] * (E + 2)
    cright = [0] * (E + 2)

    edges = list(map(int, input().split()))

    # 간선의 개수만큼 잘라냄
    for i in range(E):
        # 부모 노드번호 = 짝수
        p = edges[i * 2]
        # 자식 노드번호 = 홀수
        c = edges[i * 2 + 1]

        # p번 노드의 왼쪽자식이 없으면 왼쪽부터
        if cleft[p] == 0:
            cleft[p] = c
        # 있으면 오른쪽
        else:
            cright[p] = c

    # 문제에서 원하는 답 = N번 노드에서 순회 시작시 노드의 개수
    count = 0


    # t를 루트로 하는 서브트리 전위순회
    def preorder(t):
        global count
        # t번 노드가 존재하면
        if t != 0:
            # t번 처리(노드개수 1 증가)
            count += 1
            # t번 왼쪽 서브트리 전위순회
            preorder(cleft[t])
            # t번 오른쪽 서브트리 전위순회
            preorder(cright[t])


    preorder(N)
    print(f"#{tc} {count}")
````

### 코드 해설

- 입력으로 `E(간선 수)`와 `N(시작 노드)`가 들어옵니다.
- 먼저 `cleft/cright`를 만든 뒤,
- `preorder(N)`을 돌면서 방문할 때마다 `count += 1`을 합니다.
  - 방문 횟수 = 서브트리의 노드 개수

자주 하는 실수
- `count`를 매 테스트케이스마다 0으로 초기화하지 않으면 값이 누적됩니다.
- 시작 노드가 `N`인데 `preorder(1)`로 고정 호출하면 틀립니다.

```
    left = eval(left_child[t])
    right = eval(right_child[t])
    return apply_operator(node[t], left, right)
```


## 8. 복잡도

- 트리 만들기(`cleft/cright` 채우기): 간선이 E개면 **O(E)**
- 순회(전/중/후위): 노드 N개를 한 번씩 방문 → **O(N)**
- 수식 트리 계산(후위): 각 노드 1번 계산 → **O(N)**
- 공간: `cleft/cright/parent/tree` 배열을 쓰면 보통 **O(N)**

> 팁: 문제에서 “노드가 최대 10^5” 같은 큰 값이면, 재귀 깊이(파이썬 recursion limit)도 같이 신경 써야 합니다.


## 9. 요약(짧게)

- 트리는 **계층 구조**이고, 이진 트리는 자식이 최대 2개인 트리.
- 표현 방식은 크게 4가지: **배열 인덱스 규칙 / `cleft-cright` / `parent` / Node 클래스**  
  (문제 입력 형태와 트리 모양에 따라 골라 쓰면 됩니다.)
- 순회는 전/중/후위 3가지가 기본이고, 수식 트리는 **후위 순회로 값 계산**하면 자연스럽다.
