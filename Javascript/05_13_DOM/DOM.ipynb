학습목표
- DOM이 무엇인지 설명하고, HTML과의 관계를 이해한다.
- let과 const를 사용해 변수와 상수를 선언할 수 있다.
- querySelector와 querySelectorAll로 DOM 요소를 선택한다.
- textContent와 setAttribute로 요소 내용과 속성을 바꾼다.
- classList 속성으로 요소의 클래스를 동적으로 조작한다.
- createElement와 appendChild로 새 DOM 요소를 추가한다.
- style 프로퍼터를 사용해 요소의 스타일을 직접 변경한다.
