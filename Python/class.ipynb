Class

두 학생의 국어, 영어, 수학 점수가 각각 (90, 80, 90), (80, 70, 60) 이라 헀을 때 코드를 어떻게 작성해볼 수 있을까요?
가장 간단한 방법은변수를 6개 만들면 됩니다.

kor1, eng1, math1 = 90, 80, 90
kor2, eng2, math2 = 80, 70, 60

하지만 (국어, 영어, 수학) 이렇게 3개의 세트가 한 학생에 대한 정보를 나타내는 것에 비해, 변수를 직접 3개씩 정의하여 쓰는 것은 만족스러운코드라 할 수 없습니다.  

이런 경우에는 학생을 지칭한느 새로운 형태를 하나 정의하고, 그 형태를 이루기 위한 요소로(국어, 영어, 수학)을 정의해볼 수 있을 것입니다.  

이를 가능하게 하는 것이 바로 클래스(class)입니다.

python3에서 클래스는 다음과 같이 정의해볼 수 있습니다.

```python
class Student:
    def __init__(self, kor, eng, math):
        self.kor = kor
        self.eng = eng
        self.math = math
```

`__init__` 함수는 이 Student 클래스의 형태를 정의하는, 생성자(constructor)라 불리는 함수입니다. self를 항상 첫 번째 인자로 넣어주고, self는 학생을 지칭합니다. 인자로 넘어로는 kor, eng, math는 현재 학생의 kor, eng, math라는 값이 되어야 하기 때문에 그 값을 대입해 주는 식으로 코들르 작성하며, 이때 이러한 kor, eng, math값을 멤버 변수라고 부릅니다.  

위의 클래스를 이용하면 다음과 같이 학생 정보를 저장하는 코드를 작성해볼 수 있습니다. Student를 마치 함수를 호출 할때처럼 사용하며, `__init__`함수에 self 다음에 인자가 정의되어 순서대로 값을 적어주게 되면, 각각의 값을 갖는 하나의 학생 객체(instance)를 만들어내게 됩니다. 이 객체는 kor, eng, math 라는 이름의 멤버 변수를 갖으며, 이 값들을 조회하기 위해서는 객체이름.멤버변수이름 형태로 사용이 가능합니다.

```python
student1 = Student(90, 80, 90)
print(student1.kor)
print(student1.eng)
print(student1.math)
```

이처럼 원하는 형태의 Class를 선언하고, 해당 Class를 이용하여 원하는 객체를 만들게 되면, 원하는 형태 단위로 값을 관리할 수 있게 됩니다.

만약 멤버 변수의 이름을 k, e, m으로 정의했다면, 객체 생성 후 값을 조회할 때 이름을 k, e, m으로 사용하면 됩니다.
